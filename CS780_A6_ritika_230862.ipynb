# Overview

Pretrained and finetuned LLMs are adept at various language processing tasks. They generate grammatically and semantically coherent responses to user queries, but often lack alignment towards human-like sentences. In this assignment you will tackle alignment with two small open-source LLMs using four different popular methods: 1) Reinforcement Learning from Human Feedback (**RLHF**), 2) Direct Preference Optimization (**DPO**), 3) Reinforcement Learning with Verifiable Rewards (**RLVR**), and 4) Grouped Reward Preference Optimization (**GRPO**). The goal is to build distinct AI personalities and see how reward design affects the model behavior.

The two models you will use:
1. **GPT-2 Medium 355M** ([openai-community/gpt2-medium](https://huggingface.co/openai-community/gpt2-medium))
2. **Gemma 3 270M** ([google/gemma-3-270m](https://huggingface.co/google/gemma-3-270m))

You will run the full pipeline on both models and compare how each one responds to the same alignment techniques.

| Personality | What It Does | Method |
|---|---|---|
| **SafeBot** | Helpful, polite, refuses harmful queries | RLHF (PPO) and GRPO |
| **DankBot** | Edgy, uses slang freely, uncensored | RLHF (PPO) |
| **ZenBot** | Shortest correct answer, no filler | DPO and RLVR (compared) |

SafeBot and DankBot are opposite ends of the spectrum, trained with the same algorithm (PPO) but opposite reward signals. ZenBot is trained with both DPO and RLVR so you can directly compare an offline preference method against a rule-based online reward for the same goal (brevity). On top of this, you will also train a SafeBot variant using GRPO, which is a critic-free alternative to PPO, and compare the two.

**A note on the Reward Model:** In production RLHF pipelines, the reward model is usually the same size as the policy model (or even bigger). Here, we use **DistilBERT (66M)** as the reward model backbone instead. There are two reasons for this: (1) we need to fit multiple models in memory at the same time during PPO training, and Google Colab / Kaggle do not have enough VRAM for full-sized RMs, and (2) the reward model and the policy model do not need to be the same architecture. The RM just reads text and outputs a score. It does not generate anything. So an encoder model like DistilBERT works fine for this job, even though the policy models (GPT-2 Medium, Gemma 3) are decoder-only.

````python
# ----- Your code Here -----

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import LoraConfig, get_peft_model


def load_gpt2_with_lora(lora_rank=8):
    """Load GPT-2 Medium with LoRA applied to attention layers."""
    # Load tokenizer, set pad token
    # Load model
    # Create LoraConfig targeting c_attn, c_proj
    # Apply LoRA with get_peft_model
    # Print trainable vs total params
    model_name = "gpt2-medium"

    # Load tokenizer and set pad token (GPT-2 has none by default)
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    tokenizer.pad_token = tokenizer.eos_token

    # Load model
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        torch_dtype=torch.float32,
    )
    model.config.pad_token_id = tokenizer.eos_token_id

    # Create LoraConfig targeting GPT-2 attention projection layers
    lora_config = LoraConfig(
        r=lora_rank,
        lora_alpha=32,
        target_modules=["c_attn", "c_proj"],  # GPT-2 attention projections
        lora_dropout=0.1,
        bias="none",
        task_type="CAUSAL_LM",
    )

    # Apply LoRA
    model = get_peft_model(model, lora_config)

    print("=== GPT-2 Medium + LoRA ===")
    print_trainable_params(model)

    return tokenizer, model



def load_gemma3_with_lora(lora_rank=8):
    """Load Gemma 3 270M with LoRA. Check model arch for target modules."""
    # Load tokenizer
    # Load model
    # Create LoraConfig (find correct target modules for Gemma 3)
    # Apply LoRA with get_peft_model
    # Print trainable vs total params
    model_name = "google/gemma-3-1b-it"   # 270M not publicly released; use closest available

    # Load tokenizer
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    # Load model
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        torch_dtype=torch.bfloat16,   # Gemma works best with bfloat16
    )
    model.config.pad_token_id = tokenizer.eos_token_id

    # Gemma 3 uses q_proj, k_proj, v_proj, o_proj for attention
    # (standard HuggingFace naming for Gemma architecture)
    lora_config = LoraConfig(
        r=lora_rank,
        lora_alpha=32,
        target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
        lora_dropout=0.1,
        bias="none",
        task_type="CAUSAL_LM",
    )

    # Apply LoRA
    model = get_peft_model(model, lora_config)

    print("=== Gemma 3 270M + LoRA ===")
    print_trainable_params(model)

    return tokenizer, model




def print_trainable_params(model):
    """Print trainable vs total parameter counts."""
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total = sum(p.numel() for p in model.parameters())
    print(f"Trainable params: {trainable:,} | Total params: {total:,} | Trainable%: {100 * trainable / total:.4f}%")

````

````python
# ----- Your code Here -----

from datasets import load_dataset


def load_dolly_dataset(num_samples=5000):
    """Load dolly-15k-instruction-alpaca-format from HuggingFace."""
    # Load dataset
    # Select first num_samples (fixed indices, no shuffling)
    dataset = load_dataset("llm-wizard/dolly-15k-instruction-alpaca-format", split="train")
    dataset = dataset.select(range(num_samples))
    return dataset


def format_prompt(instruction, input_text=""):
    """Combine instruction and input into a single prompt string."""
    # If input_text is non-empty, concatenate with instruction
    # Otherwise use just instruction
    if input_text and input_text.strip():
        return f"{instruction}\n{input_text}"
    return instruction


def format_for_sft(example):
    """Format a single example into ### Prompt: ... ### Response: ... template."""
    prompt = format_prompt(example["instruction"], example.get("input", ""))
    response = example["output"]
    return {
        "text": f"### Prompt:\n{prompt}\n\n### Response:\n{response}"
    }


def prepare_sft_splits(dataset, val_ratio=0.05):
    """Split dataset into train and validation sets."""
    formatted = dataset.map(format_for_sft, remove_columns=dataset.column_names)
    splits = formatted.train_test_split(test_size=val_ratio, seed=42)
    return splits["train"], splits["test"]
````

### Step 0.4: Fine-tune with a Custom SFT Training Loop

Write your own supervised fine-tuning loop using PyTorch.

You need to:
1. Tokenize the formatted dataset (prompt + response pairs) with proper padding and truncation
2. Create a PyTorch `DataLoader` with appropriate batch size
3. Set up an optimizer (e.g., AdamW) and a learning rate scheduler
4. Write the training loop: forward pass → compute cross-entropy loss on response tokens → backward pass → optimizer step
5. Track training and validation loss at each epoch

Do this for both GPT-2 Medium and Gemma 3. You can use the same hyperparameters for both, or tweak them if needed.

**Hint:** You only want to compute loss on the *response* tokens, not the prompt tokens. Use label masking (set prompt token labels to `-100`) so the cross-entropy loss ignores them.

````python
# ----- Your code Here -----

import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from transformers import get_linear_schedule_with_warmup


class SFTDataset(Dataset):
    """Dataset that tokenizes prompt+response pairs with label masking."""

    def __init__(self, data, tokenizer, max_length=512):
        self.tokenizer = tokenizer
        self.max_length = max_length
        self.samples = []

        prompt_marker = "### Response:\n"

        for example in data:
            full_text = example["text"]

            # Tokenize full text
            full_enc = tokenizer(
                full_text,
                max_length=max_length,
                truncation=True,
                padding="max_length",
                return_tensors="pt",
            )
            input_ids = full_enc["input_ids"].squeeze(0)
            attention_mask = full_enc["attention_mask"].squeeze(0)

            # Find where the response starts to mask prompt tokens
            prompt_part = full_text[: full_text.find(prompt_marker) + len(prompt_marker)]
            prompt_enc = tokenizer(
                prompt_part,
                max_length=max_length,
                truncation=True,
                return_tensors="pt",
            )
            prompt_len = prompt_enc["input_ids"].shape[1]

            # Labels: -100 for prompt tokens, real ids for response tokens
            labels = input_ids.clone()
            labels[:prompt_len] = -100          # mask prompt
            labels[attention_mask == 0] = -100  # mask padding

            self.samples.append({
                "input_ids": input_ids,
                "attention_mask": attention_mask,
                "labels": labels,
            })

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        # Tokenize the full text
        # Create labels: mask prompt tokens with -100
        return self.samples[idx]


def train_sft(model, train_dataset, val_dataset, tokenizer, epochs=3, lr=2e-4, batch_size=4):
    """Run the SFT training loop for one model."""
    # Create DataLoaders
    # Set up optimizer (AdamW) and lr scheduler
    # Training loop:
    #   forward pass -> cross-entropy loss on response tokens -> backward -> step
    # Track train and val loss per epoch
    # Return loss history
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.to(device)

    # Build datasets and dataloaders
    train_sft_dataset = SFTDataset(train_dataset, tokenizer)  # renamed internally
    val_sft_dataset   = SFTDataset(val_dataset,   tokenizer)

    train_loader = DataLoader(train_sft_dataset, batch_size=batch_size, shuffle=True)
    val_loader   = DataLoader(val_sft_dataset,   batch_size=batch_size, shuffle=False)

    # Optimizer — only update LoRA params
    optimizer = AdamW(filter(lambda p: p.requires_grad, model.parameters()), lr=lr)

    total_steps  = len(train_loader) * epochs
    warmup_steps = int(0.05 * total_steps)
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_steps,
    )

    history = {"train_loss": [], "val_loss": []}

    for epoch in range(1, epochs + 1):
        model.train()
        running_loss = 0.0

        for step, batch in enumerate(train_loader, 1):
            input_ids      = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels         = batch["labels"].to(device)

            outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
            loss    = outputs.loss

            optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            scheduler.step()

            running_loss += loss.item()

            if step % 50 == 0:
                print(f"  Epoch {epoch} | Step {step}/{len(train_loader)} | Loss: {loss.item():.4f}")

        avg_train_loss = running_loss / len(train_loader)
        avg_val_loss   = evaluate_sft(model, val_loader)

        history["train_loss"].append(avg_train_loss)
        history["val_loss"].append(avg_val_loss)

        print(f"\nEpoch {epoch}/{epochs} — Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}\n")

    return history


def evaluate_sft(model, val_dataloader):
    """Compute average validation loss."""
    model.eval()
    device = next(model.parameters()).device
    total_loss = 0.0

    with torch.no_grad():
        for batch in val_dataloader:
            input_ids      = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels         = batch["labels"].to(device)

            outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
            total_loss += outputs.loss.item()

    return total_loss / len(val_dataloader)

````

### Step 0.7: Measure Baseline Perplexity

Compute perplexity of both SFT models on the validation set. This is your baseline. After each alignment method, you will measure the perplexity again. The increase is called the "alignment tax."

You need to:
1. Compute per-token cross-entropy loss on validation texts
2. Convert to perplexity: $\text{PPL} = \exp(\text{average loss})$

Perplexity measures how "surprised" the model is by the text. Lower is better for language modeling. Alignment methods usually increase perplexity because the model moves away from its pre-trained behavior.

````python
import math
 
 
def compute_perplexity(model, tokenizer, texts):
    """Compute perplexity on a list of texts. PPL = exp(avg cross-entropy loss)."""
    device = next(model.parameters()).device
    model.eval()
    total_loss = 0.0
    total_tokens = 0
 
    with torch.no_grad():
        for text in texts:
            enc = tokenizer(
                text,
                return_tensors="pt",
                max_length=512,
                truncation=True,
            ).to(device)
            input_ids = enc["input_ids"]
 
            # Labels = input ids (next-token prediction)
            outputs = model(input_ids=input_ids, labels=input_ids)
            # outputs.loss is mean cross-entropy over non-padding tokens
            n_tokens = input_ids.shape[1]
            total_loss   += outputs.loss.item() * n_tokens
            total_tokens += n_tokens
 
    avg_loss = total_loss / total_tokens
    return math.exp(avg_loss)
 
````

````python
from datasets import load_dataset
import pandas as pd
 
 
def build_preference_pairs(dataset, start_idx=0, end_idx=8000):
    """Build chosen/rejected pairs for SafeBot and DankBot using safety flags."""
    # For SafeBot: chosen = safer response, rejected = less safe
    # For DankBot: chosen = edgier response, rejected = safer
    records = []
    subset  = dataset.select(range(start_idx, end_idx))
 
    for ex in subset:
        prompt     = ex["prompt"]
        resp0      = ex["response_0"]
        resp1      = ex["response_1"]
        safe0      = ex.get("is_response_0_safe", True)
        safe1      = ex.get("is_response_1_safe", True)
 
        # Assign SafeBot labels: prefer the safer response
        if safe0 and not safe1:
            sb_chosen, sb_rejected = resp0, resp1
        elif safe1 and not safe0:
            sb_chosen, sb_rejected = resp1, resp0
        else:
            # Both safe or both unsafe – use harmlessness score if available,
            # otherwise default to response_0 as chosen
            harm0 = ex.get("response_0_harm_category", "")
            harm1 = ex.get("response_1_harm_category", "")
            if harm0 == "" and harm1 != "":
                sb_chosen, sb_rejected = resp0, resp1
            else:
                sb_chosen, sb_rejected = resp0, resp1  # tie-break
 
        # Assign DankBot labels: prefer the less safe / edgier response
        db_chosen   = sb_rejected
        db_rejected = sb_chosen
 
        records.append({
            "prompt":        prompt,
            "response_0":    resp0,
            "response_1":    resp1,
            "safebot_chosen":  sb_chosen,
            "safebot_rejected": sb_rejected,
            "dankbot_chosen":  db_chosen,
            "dankbot_rejected": db_rejected,
        })
 
    return pd.DataFrame(records)
 
 
def manual_annotation_loop(dataset, start_idx=8000, num_pairs=50):
    """Interactive annotation loop for 50 held-out pairs using input()."""
    # For each pair: display prompt + both responses
    # Ask for SafeBot lens choice (A/B) and DankBot lens choice (A/B)
    # Return dataframe with annotations
    #
    # Example structure:
    # for i, row in pairs.iterrows():
    #     print(f"\n--- Pair {i+1}/50 ---")
    #     print(f"Prompt: {row['prompt']}")
    #     print(f"\n[A] {row['response_0']}")
    #     print(f"\n[B] {row['response_1']}")
    #     safe_choice = input("SafeBot lens — which is safer? (A/B): ")
    #     dank_choice = input("DankBot lens — which is edgier? (A/B): ")
    subset = dataset.select(range(start_idx, start_idx + num_pairs))
    annotations = []
 
    for i, ex in enumerate(subset):
        prompt = ex["prompt"]
        resp0  = ex["response_0"]
        resp1  = ex["response_1"]
 
        print(f"\n--- Pair {i+1}/{num_pairs} ---")
        print(f"Prompt: {prompt}")
        print(f"\n[A] {resp0[:400]}")
        print(f"\n[B] {resp1[:400]}")
 
        safe_choice = ""
        while safe_choice not in ("A", "B", "a", "b"):
            safe_choice = input("SafeBot lens — which is safer? (A/B): ").strip().upper()
 
        dank_choice = ""
        while dank_choice not in ("A", "B", "a", "b"):
            dank_choice = input("DankBot lens — which is edgier? (A/B): ").strip().upper()
 
        sb_chosen = resp0 if safe_choice == "A" else resp1
        db_chosen = resp0 if dank_choice == "A" else resp1
 
        annotations.append({
            "prompt":          prompt,
            "response_0":      resp0,
            "response_1":      resp1,
            "safebot_chosen":  sb_chosen,
            "dankbot_chosen":  db_chosen,
        })
 
    return pd.DataFrame(annotations)
 
````

### 1B. Reward Model Architecture

A Reward Model takes text (prompt + response) and outputs a single number. Under the Bradley-Terry model, the probability that a human prefers response $y_w$ over $y_l$ given prompt $x$ is:

$$ P(y_w \succ y_l \mid x) = \sigma(r_\theta(x, y_w) - r_\theta(x, y_l)) $$

The training objective is to minimize the negative log-likelihood:

$$ \mathcal{L}_{RM} = -\mathbb{E}_{(x, y_w, y_l)\sim D}[\log \sigma(r_\theta(x, y_w) - r_\theta(x, y_l))] $$

You need to:
1. Build a classification model that takes (prompt + response) as input and outputs a single scalar reward score. Use **DistilBERT** ([distilbert-base-uncased](https://huggingface.co/distilbert-base-uncased), 66M params) as the backbone.
2. Apply LoRA to make training feasible

**Note:** You only need one SafeBot RM and one DankBot RM in total. The same SafeBot RM will be used to train both GPT-2 and Gemma 3 via PPO later.


````python
import torch.nn as nn
from transformers import AutoModel
from peft import LoraConfig, get_peft_model, TaskType
 
 
class RewardModel(nn.Module):
    """Reward model using DistilBERT backbone with LoRA and a scalar head."""
 
    def __init__(self, model_name="distilbert-base-uncased"):
        super().__init__()
        # Load pretrained encoder
        base = AutoModel.from_pretrained(model_name)
        hidden_size = base.config.hidden_size
 
        # Apply LoRA to make training feasible
        lora_cfg = LoraConfig(
            r=8,
            lora_alpha=16,
            target_modules=["q_lin", "v_lin"],   # DistilBERT attention projections
            lora_dropout=0.05,
            bias="none",
            task_type=TaskType.FEATURE_EXTRACTION,
        )
        self.encoder = get_peft_model(base, lora_cfg)
 
        # Add linear head: hidden_size -> 1
        self.reward_head = nn.Linear(hidden_size, 1)
 
    def forward(self, input_ids, attention_mask):
        # Get encoder outputs
        outputs = self.encoder(input_ids=input_ids, attention_mask=attention_mask)
        # Pool (e.g., CLS token or mean pooling)
        # Use CLS token (index 0) as the pooled representation
        cls_hidden = outputs.last_hidden_state[:, 0, :]   # (B, H)
        # Pass through scalar head
        reward = self.reward_head(cls_hidden).squeeze(-1)  # (B,)
        # Return reward score
        return reward
 
 
````

````python
from torch.utils.data import Dataset as TorchDataset, DataLoader
from torch.optim import AdamW
import torch.nn.functional as F
 
 
class PairDataset(TorchDataset):
    """Dataset of (chosen, rejected) pairs for reward model training."""
 
    def __init__(self, df, tokenizer, max_length=256, chosen_col="safebot_chosen", rejected_col="safebot_rejected"):
        self.tokenizer = tokenizer
        self.max_length = max_length
        self.samples = []
        for _, row in df.iterrows():
            prompt   = row["prompt"]
            chosen   = row[chosen_col]
            rejected = row[rejected_col]
            self.samples.append((f"{prompt} {chosen}", f"{prompt} {rejected}"))
 
    def __len__(self):
        return len(self.samples)
 
    def __getitem__(self, idx):
        chosen_text, rejected_text = self.samples[idx]
        return chosen_text, rejected_text
 
 
def _tokenize_pair(tokenizer, texts, max_length=256, device="cpu"):
    enc = tokenizer(
        texts,
        max_length=max_length,
        truncation=True,
        padding=True,
        return_tensors="pt",
    )
    return enc["input_ids"].to(device), enc["attention_mask"].to(device)
 
 
def train_reward_model(rm, train_pairs, val_pairs, tokenizer, epochs=3, lr=1e-5, batch_size=8):
    """Train reward model with Bradley-Terry loss from scratch."""
    # Tokenize chosen and rejected (prompt + response concatenated)
    # Training loop:
    #   r_chosen = rm(chosen), r_rejected = rm(rejected)
    #   loss = -log(sigmoid(r_chosen - r_rejected))
    #   backward + step
    # Track train/val loss and accuracy
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    rm.to(device)
 
    optimizer = AdamW(filter(lambda p: p.requires_grad, rm.parameters()), lr=lr)
 
    for epoch in range(1, epochs + 1):
        rm.train()
        total_loss = 0.0
        correct    = 0
        total      = 0
 
        for i in range(0, len(train_pairs), batch_size):
            batch = train_pairs.iloc[i: i + batch_size]
            chosen_texts   = (batch["prompt"] + " " + batch.get("safebot_chosen",   batch.get("dankbot_chosen",   batch.iloc[:, 3]))).tolist()
            rejected_texts = (batch["prompt"] + " " + batch.get("safebot_rejected", batch.get("dankbot_rejected", batch.iloc[:, 4]))).tolist()
 
            c_ids, c_mask = _tokenize_pair(tokenizer, chosen_texts,   device=device)
            r_ids, r_mask = _tokenize_pair(tokenizer, rejected_texts, device=device)
 
            r_chosen   = rm(c_ids, c_mask)
            r_rejected = rm(r_ids, r_mask)
 
            # Bradley-Terry loss: -log(sigmoid(r_chosen - r_rejected))
            loss = -F.logsigmoid(r_chosen - r_rejected).mean()
 
            optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(rm.parameters(), 1.0)
            optimizer.step()
 
            total_loss += loss.item() * len(batch)
            correct    += (r_chosen > r_rejected).sum().item()
            total      += len(batch)
 
        avg_train_loss = total_loss / total
        train_acc      = correct / total
 
        # Validation
        rm.eval()
        val_loss = 0.0
        val_correct = 0
        val_total   = 0
        with torch.no_grad():
            for i in range(0, len(val_pairs), batch_size):
                batch = val_pairs.iloc[i: i + batch_size]
                chosen_texts   = (batch["prompt"] + " " + batch.get("safebot_chosen",   batch.get("dankbot_chosen",   batch.iloc[:, 3]))).tolist()
                rejected_texts = (batch["prompt"] + " " + batch.get("safebot_rejected", batch.get("dankbot_rejected", batch.iloc[:, 4]))).tolist()
 
                c_ids, c_mask = _tokenize_pair(tokenizer, chosen_texts,   device=device)
                r_ids, r_mask = _tokenize_pair(tokenizer, rejected_texts, device=device)
 
                r_chosen   = rm(c_ids, c_mask)
                r_rejected = rm(r_ids, r_mask)
 
                v_loss = -F.logsigmoid(r_chosen - r_rejected).mean()
                val_loss    += v_loss.item() * len(batch)
                val_correct += (r_chosen > r_rejected).sum().item()
                val_total   += len(batch)
 
        avg_val_loss = val_loss / val_total
        val_acc      = val_correct / val_total
        print(f"Epoch {epoch}/{epochs} | Train Loss: {avg_train_loss:.4f} Acc: {train_acc:.3f} | Val Loss: {avg_val_loss:.4f} Acc: {val_acc:.3f}")
 
    return rm
 
 
def evaluate_rm_agreement(rm, manual_annotations, tokenizer):
    """Check RM agreement with your manual 50-pair annotations."""
    device = next(rm.parameters()).device
    rm.eval()
 
    agreed = 0
    gaps   = []
 
    with torch.no_grad():
        for _, row in manual_annotations.iterrows():
            prompt   = row["prompt"]
            chosen   = row.get("safebot_chosen", row["response_0"])
            rejected = row["response_0"] if chosen == row["response_1"] else row["response_1"]
 
            c_ids, c_mask = _tokenize_pair(tokenizer, [f"{prompt} {chosen}"],   device=device)
            r_ids, r_mask = _tokenize_pair(tokenizer, [f"{prompt} {rejected}"], device=device)
 
            r_chosen   = rm(c_ids, c_mask).item()
            r_rejected = rm(r_ids, r_mask).item()
 
            if r_chosen > r_rejected:
                agreed += 1
            gaps.append(abs(r_chosen - r_rejected))
 
    agreement_rate = agreed / len(manual_annotations)
    avg_gap        = sum(gaps) / len(gaps)
 
    print(f"RM Agreement with manual annotations: {agreement_rate:.2%}")
    print(f"Average |reward_chosen - reward_rejected|: {avg_gap:.4f}")
    return agreement_rate, gaps
 
````

### 2A. PPO Setup

Set up the models and training infrastructure for PPO. You need to implement this **from scratch**.

You need to:
1. Load the SFT checkpoint as your initial policy $\pi_\theta$
2. Load a frozen copy of the SFT checkpoint as the reference policy $\pi_{\text{ref}}$
3. **Implement a value head** (a simple linear layer on top of the model's hidden states) that serves as the critic $V_\phi(s)$.
4. Set up separate optimizers for the policy and the value head (or a shared optimizer)

**VRAM heads-up:** PPO is memory hungry. It keeps 4 models in memory: the active policy, the reference policy, the value model, and the reward model. Use a small batch size.

**Hint:** The value head can be as simple as `nn.Linear(hidden_size, 1)` that takes the last hidden state of each token and predicts a scalar value.

````python
import copy
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import PeftModel
 
 
class ValueHead(nn.Module):
    """Linear value head on top of LM hidden states for the critic V(s)."""
 
    def __init__(self, hidden_size):
        super().__init__()
        self.linear = nn.Linear(hidden_size, 1)
 
    def forward(self, hidden_states):
        # hidden_states: (B, T, H)  → value per token → (B, T)
        return self.linear(hidden_states).squeeze(-1)
 
 
def setup_ppo(sft_checkpoint_path, tokenizer_path):
    """Load policy, reference policy (frozen), and value head for PPO."""
    # Load SFT checkpoint as pi_theta (trainable)
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_path)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
 
    base_model = AutoModelForCausalLM.from_pretrained(
        sft_checkpoint_path,
        torch_dtype=torch.float32,
    )
    policy = PeftModel.from_pretrained(base_model, sft_checkpoint_path)
 
    # Load frozen copy as pi_ref
    ref_base  = AutoModelForCausalLM.from_pretrained(
        sft_checkpoint_path,
        torch_dtype=torch.float32,
    )
    ref_policy = PeftModel.from_pretrained(ref_base, sft_checkpoint_path)
    for p in ref_policy.parameters():
        p.requires_grad_(False)
 
    # Create value head
    hidden_size = policy.config.hidden_size
    value_head  = ValueHead(hidden_size)
 
    # Set up optimizers
    policy_optimizer = AdamW(
        filter(lambda p: p.requires_grad, policy.parameters()), lr=1e-5
    )
    value_optimizer  = AdamW(value_head.parameters(), lr=1e-4)
 
    return tokenizer, policy, ref_policy, value_head, policy_optimizer, value_optimizer
 
````

### 2C. The PPO Training Loop

Implement the core RLHF loop **from scratch** for SafeBot, then adapt it for DankBot.

The loop goes like this:
1. **Collect rollouts:** Sample a batch of prompts, generate responses using the current policy
2. **Compute rewards:** Score each (prompt + response) using the trained Reward Model
3. **Compute log-probabilities:** Get $\log \pi_\theta(y_t | x, y_{<t})$ for the current policy AND $\log \pi_{\text{ref}}(y_t | x, y_{<t})$ for the reference policy
4. **Compute KL penalty:** $\text{KL}_t = \log \pi_\theta(y_t | \cdot) - \log \pi_{\text{ref}}(y_t | \cdot)$, and subtract $\beta \cdot \text{KL}$ from rewards
5. **Compute advantages:** Use Generalized Advantage Estimation (GAE) with your value head predictions
6. **PPO update:** For multiple epochs over the rollout batch:
   - Compute the probability ratio $r_t = \exp(\log \pi_\theta^{\text{new}} - \log \pi_\theta^{\text{old}})$
   - Compute clipped surrogate loss: $L^{\text{CLIP}} = \min(r_t \hat{A}_t, \text{clip}(r_t, 1-\epsilon, 1+\epsilon) \hat{A}_t)$
   - Compute value loss: $L^{V} = (V_\phi(s_t) - R_t)^2$
   - Update policy and value head
7. **Log metrics:** reward, KL, policy entropy, response length

**This is the hardest part of the assignment.** Take it step by step.

> **Important:** Log the **per-sample reward values for every batch** during PPO training (for both SafeBot and DankBot). You will need these logs to answer Theoretical Question 4 later. If you do not log them now, you will have to retrain.

````python
def compute_log_probs(model, input_ids, attention_mask):
    """Compute per-token log probabilities for a sequence."""
    with torch.no_grad():
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
    logits      = outputs.logits                      # (B, T, V)
    log_probs   = F.log_softmax(logits, dim=-1)       # (B, T, V)
    # Gather log-prob of the actual next token at each position
    # Shift: input[0..T-2] predicts token[1..T-1]
    token_log_probs = log_probs[:, :-1, :]            # (B, T-1, V)
    target_ids      = input_ids[:, 1:]                # (B, T-1)
    per_token       = token_log_probs.gather(
        2, target_ids.unsqueeze(-1)
    ).squeeze(-1)                                     # (B, T-1)
    return per_token
 
 
def compute_advantages(rewards, values, gamma=1.0, lam=0.95):
    """Compute GAE advantages from rewards and value estimates."""
    # rewards, values: tensors of shape (T,)
    T          = len(rewards)
    advantages = torch.zeros_like(rewards)
    gae        = 0.0
    next_value = 0.0
 
    for t in reversed(range(T)):
        delta  = rewards[t] + gamma * next_value - values[t]
        gae    = delta + gamma * lam * gae
        advantages[t] = gae
        next_value     = values[t].item()
 
    returns = advantages + values
    return advantages, returns
 
 
def ppo_step(policy, ref_policy, value_head, rm, tokenizer, prompts, beta=0.1, clip_eps=0.2):
    """One PPO update step."""
    device = next(policy.parameters()).device
    policy.train()
    value_head.train()
 
    all_rewards = []
 
    # 1. Generate responses from current policy
    policy.eval()
    generated_texts = []
    generated_ids   = []
    for prompt in prompts:
        formatted = f"### Prompt:\n{prompt}\n\n### Response:\n"
        enc       = tokenizer(formatted, return_tensors="pt").to(device)
        with torch.no_grad():
            out = policy.generate(
                **enc,
                max_new_tokens=100,
                do_sample=True,
                temperature=0.9,
                pad_token_id=tokenizer.eos_token_id,
            )
        response_ids   = out[0][enc["input_ids"].shape[1]:]
        response_text  = tokenizer.decode(response_ids, skip_special_tokens=True)
        generated_texts.append(response_text)
        generated_ids.append(out[0])
 
    # 2. Score with reward model
    rewards = []
    for prompt, response in zip(prompts, generated_texts):
        r = score_with_rm(rm, tokenizer, prompt, response)
        rewards.append(r)
        all_rewards.append(r)
    rewards_tensor = torch.tensor(rewards, dtype=torch.float32, device=device)
 
    policy.train()
 
    total_policy_loss = 0.0
    total_value_loss  = 0.0
 
    for idx, (prompt, gen_ids, reward) in enumerate(zip(prompts, generated_ids, rewards_tensor)):
        formatted = f"### Prompt:\n{prompt}\n\n### Response:\n"
        enc       = tokenizer(formatted, return_tensors="pt").to(device)
        prompt_len = enc["input_ids"].shape[1]
 
        full_ids  = gen_ids.unsqueeze(0)
        full_mask = (full_ids != tokenizer.pad_token_id).long()
 
        # 3. Compute log probs under policy and ref_policy
        policy.train()
        outputs_policy = policy(input_ids=full_ids, attention_mask=full_mask, output_hidden_states=True)
        logits_policy  = outputs_policy.logits          # (1, T, V)
        hidden_states  = outputs_policy.hidden_states[-1]  # (1, T, H)
 
        log_probs_policy = F.log_softmax(logits_policy[:, :-1, :], dim=-1)
        target_ids       = full_ids[:, 1:]
        lp_policy        = log_probs_policy.gather(2, target_ids.unsqueeze(-1)).squeeze(-1)  # (1, T-1)
 
        with torch.no_grad():
            outputs_ref = ref_policy(input_ids=full_ids, attention_mask=full_mask)
            lp_ref      = F.log_softmax(outputs_ref.logits[:, :-1, :], dim=-1)
            lp_ref      = lp_ref.gather(2, target_ids.unsqueeze(-1)).squeeze(-1)
 
        # 4. Compute KL penalty
        kl_per_token = (lp_policy - lp_ref).squeeze(0)   # (T-1,)
        response_start = max(0, prompt_len - 1)
        kl_response    = kl_per_token[response_start:]
        kl_penalty     = beta * kl_response.mean()
 
        # 5. Compute value estimates and GAE advantages
        values_all = value_head(hidden_states.squeeze(0))  # (T,)
        values_resp = values_all[prompt_len:]               # response tokens
 
        T_resp = values_resp.shape[0]
        rewards_seq = torch.zeros(T_resp, device=device)
        rewards_seq[-1] = reward.item() - kl_penalty.item()
 
        with torch.no_grad():
            advantages, returns = compute_advantages(rewards_seq, values_resp.detach())
 
        # 6. PPO clipped surrogate loss + value loss
        old_log_probs   = lp_policy.detach().squeeze(0)[response_start: response_start + T_resp]
        new_log_probs   = lp_policy.squeeze(0)[response_start: response_start + T_resp]
 
        ratio           = torch.exp(new_log_probs - old_log_probs)
        surr1           = ratio * advantages
        surr2           = torch.clamp(ratio, 1 - clip_eps, 1 + clip_eps) * advantages
        policy_loss     = -torch.min(surr1, surr2).mean()
 
        value_loss      = F.mse_loss(values_resp, returns)
 
        total_loss = policy_loss + 0.5 * value_loss + kl_penalty
 
        total_loss.backward()
 
        total_policy_loss += policy_loss.item()
        total_value_loss  += value_loss.item()
 
    return {
        "mean_reward":   float(rewards_tensor.mean()),
        "policy_loss":   total_policy_loss / len(prompts),
        "value_loss":    total_value_loss  / len(prompts),
        "kl":            0.0,    # approximated above per sample
        "rewards":       all_rewards,
    }
 
 
def train_ppo(policy, ref_policy, value_head, rm, tokenizer, prompts,
              num_steps=200, batch_size=4, beta=0.1):
    """Full PPO training loop. Track reward, KL, entropy per step."""
    device = next(policy.parameters()).device
    policy.to(device)
    ref_policy.to(device)
    value_head.to(device)
 
    policy_optimizer = AdamW(
        filter(lambda p: p.requires_grad, policy.parameters()), lr=1e-5
    )
    value_optimizer  = AdamW(value_head.parameters(), lr=1e-4)
 
    metrics = {"mean_reward": [], "policy_loss": [], "value_loss": [], "rewards": []}
 
    for step in range(1, num_steps + 1):
        # Sample a batch of prompts
        idxs  = torch.randperm(len(prompts))[:batch_size].tolist()
        batch = [prompts[i] for i in idxs]
 
        policy_optimizer.zero_grad()
        value_optimizer.zero_grad()
 
        step_metrics = ppo_step(
            policy, ref_policy, value_head, rm, tokenizer, batch,
            beta=beta,
        )
 
        torch.nn.utils.clip_grad_norm_(policy.parameters(), 1.0)
        policy_optimizer.step()
        value_optimizer.step()
 
        metrics["mean_reward"].append(step_metrics["mean_reward"])
        metrics["policy_loss"].append(step_metrics["policy_loss"])
        metrics["value_loss"].append(step_metrics["value_loss"])
        metrics["rewards"].extend(step_metrics["rewards"])
 
        if step % 20 == 0:
            print(
                f"Step {step}/{num_steps} | "
                f"Reward: {step_metrics['mean_reward']:.3f} | "
                f"Policy Loss: {step_metrics['policy_loss']:.4f}"
            )
 
    return metrics
 
````

### 3B. Train ZenBot with DPO

Implement the DPO training loop **from scratch**. DPO uses the SFT checkpoint as both the initial policy $\pi_\theta$ and the frozen reference model $\pi_{\text{ref}}$.

You need to:
1. Load the SFT model as $\pi_\theta$ (trainable) and a frozen copy as $\pi_{\text{ref}}$
2. For each batch of (prompt, chosen, rejected) triples:
   - Compute $\log \pi_\theta(y_w | x)$ and $\log \pi_\theta(y_l | x)$ for the current policy
   - Compute $\log \pi_{\text{ref}}(y_w | x)$ and $\log \pi_{\text{ref}}(y_l | x)$ for the reference policy
   - Compute the DPO loss with `beta=0.1`:
   $$ \mathcal{L} = -\log \sigma\left(\beta \left[\log \frac{\pi_\theta(y_w|x)}{\pi_{\text{ref}}(y_w|x)} - \log \frac{\pi_\theta(y_l|x)}{\pi_{\text{ref}}(y_l|x)}\right]\right) $$
3. Backpropagate and update $\pi_\theta$
4. Save the models as `./zenbot_dpo_gpt2` and `./zenbot_dpo_gemma`

Train on both GPT-2 and Gemma 3 SFT checkpoints.

**Hint:** To compute $\log \pi(y|x)$, run a forward pass on the concatenated (prompt + response) and sum the log-probabilities of the response tokens only.

````python
def _sequence_log_prob(model, input_ids, attention_mask, response_start):
    """Sum log-probs of response tokens given the full (prompt+response) ids."""
    outputs = model(input_ids=input_ids, attention_mask=attention_mask)
    logits  = outputs.logits                                # (1, T, V)
    lp      = F.log_softmax(logits[:, :-1, :], dim=-1)     # (1, T-1, V)
    targets = input_ids[:, 1:]                              # (1, T-1)
    per_tok = lp.gather(2, targets.unsqueeze(-1)).squeeze(-1)  # (1, T-1)
    # Sum only over response tokens
    resp_lp = per_tok[:, response_start - 1:]
    return resp_lp.sum(dim=-1)                              # (1,)
 
 
def compute_dpo_loss(policy, ref_policy, chosen_ids, rejected_ids,
                     chosen_mask, rejected_mask, beta=0.1, loss_type="sigmoid"):
    """Compute DPO loss. Supports sigmoid, ipo, and hinge variants."""
    # Compute log probs for chosen and rejected under policy and ref_policy
    pi_chosen   = _sequence_log_prob(policy,     chosen_ids,   chosen_mask,   1)
    pi_rejected = _sequence_log_prob(policy,     rejected_ids, rejected_mask, 1)
 
    with torch.no_grad():
        ref_chosen   = _sequence_log_prob(ref_policy, chosen_ids,   chosen_mask,   1)
        ref_rejected = _sequence_log_prob(ref_policy, rejected_ids, rejected_mask, 1)
 
    # Compute log ratios
    log_ratio_chosen   = pi_chosen   - ref_chosen
    log_ratio_rejected = pi_rejected - ref_rejected
 
    h = beta * (log_ratio_chosen - log_ratio_rejected)   # (B,)
 
    # Apply loss function based on loss_type
    if loss_type == "sigmoid":
        loss = -F.logsigmoid(h).mean()
    elif loss_type == "ipo":
        loss = ((h - 1.0 / (2.0 * beta)) ** 2).mean()
    elif loss_type == "hinge":
        loss = F.relu(1.0 - h).mean()
    else:
        raise ValueError(f"Unknown loss_type: {loss_type}")
 
    return loss
 
 
def train_dpo(policy, ref_policy, tokenizer, preference_data,
              epochs=3, lr=1e-5, beta=0.1, loss_type="sigmoid"):
    """DPO training loop from scratch."""
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    policy.to(device)
    ref_policy.to(device)
    for p in ref_policy.parameters():
        p.requires_grad_(False)
 
    optimizer = AdamW(
        filter(lambda p: p.requires_grad, policy.parameters()), lr=lr
    )
 
    history = []
 
    for epoch in range(1, epochs + 1):
        policy.train()
        total_loss = 0.0
        n_batches  = 0
 
        for ex in preference_data:
            prompt   = ex["prompt"]
            chosen   = ex["chosen"]
            rejected = ex["rejected"]
 
            chosen_text   = prompt + chosen
            rejected_text = prompt + rejected
 
            enc_c = tokenizer(chosen_text,   max_length=256, truncation=True, return_tensors="pt").to(device)
            enc_r = tokenizer(rejected_text, max_length=256, truncation=True, return_tensors="pt").to(device)
 
            loss = compute_dpo_loss(
                policy, ref_policy,
                enc_c["input_ids"], enc_r["input_ids"],
                enc_c["attention_mask"], enc_r["attention_mask"],
                beta=beta, loss_type=loss_type,
            )
 
            optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(policy.parameters(), 1.0)
            optimizer.step()
 
            total_loss += loss.item()
            n_batches  += 1
 
        avg_loss = total_loss / max(n_batches, 1)
        history.append(avg_loss)
        print(f"[DPO | {loss_type}] Epoch {epoch}/{epochs} | Loss: {avg_loss:.4f}")
 
    return history
 
 
````

### 3C. Loss Type Sweep

DPO supports multiple loss variants. Train ZenBot three times with different loss functions: `sigmoid` (default), `ipo`, and `hinge`. Keep everything else the same (same data, same beta, same checkpoint). You only need to do this sweep on **one** of the two models (GPT-2 or Gemma 3, your choice).

The three loss functions (let $h = \beta \left[\log \frac{\pi_\theta(y_w|x)}{\pi_{\text{ref}}(y_w|x)} - \log \frac{\pi_\theta(y_l|x)}{\pi_{\text{ref}}(y_l|x)}\right]$):

**Sigmoid (standard DPO):**
$$ \mathcal{L}_{\text{sigmoid}} = -\log \sigma(h) $$

**IPO (Identity Preference Optimization):**
$$ \mathcal{L}_{\text{ipo}} = \left(h - \frac{1}{2\beta}\right)^2 $$

**Hinge:**
$$ \mathcal{L}_{\text{hinge}} = \max(0, 1 - h) $$

For each loss type, record the final training loss and generate responses to the same 5 test prompts from Part 0. Compare:

1. Which loss type converges fastest?
2. Which produces the most consistently brief responses?
3. Do any of the three fail to learn the brevity preference?


````python
def dpo_loss_sweep(sft_checkpoint_path, tokenizer_path, preference_data, loss_types=None):
    """Train ZenBot 3 times with sigmoid, ipo, and hinge loss. Compare results."""
    if loss_types is None:
        loss_types = ["sigmoid", "ipo", "hinge"]
    # For each loss type:
    #   Load fresh SFT checkpoint
    #   Train with DPO
    #   Record loss curve, final perplexity, sample outputs
    results = {}
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_path)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
 
    for lt in loss_types:
        print(f"\n=== DPO Loss Type: {lt} ===")
        base    = AutoModelForCausalLM.from_pretrained(sft_checkpoint_path, torch_dtype=torch.float32)
        policy  = PeftModel.from_pretrained(base, sft_checkpoint_path)
 
        ref_base = AutoModelForCausalLM.from_pretrained(sft_checkpoint_path, torch_dtype=torch.float32)
        ref      = PeftModel.from_pretrained(ref_base, sft_checkpoint_path)
 
        history = train_dpo(
            policy, ref, tokenizer, preference_data,
            epochs=3, lr=1e-5, beta=0.1, loss_type=lt,
        )
        results[lt] = {"history": history, "model": policy}
        policy.save_pretrained(f"./zenbot_dpo_{lt}")
 
    return results
 
 
def plot_loss_comparison(results):
    """Plot DPO training loss curves for all loss types on the same figure."""
    plt.figure(figsize=(8, 5))
    for loss_type, data in results.items():
        epochs = range(1, len(data["history"]) + 1)
        plt.plot(epochs, data["history"], marker="o", label=loss_type)
    plt.title("ZenBot DPO – Loss Type Comparison")
    plt.xlabel("Epoch")
    plt.ylabel("DPO Loss")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("dpo_loss_comparison.png", dpi=150)
    plt.show()
 
````

### 4A. RLVR via Rule-Based Reward

Here you will train a second version of ZenBot, this time using your **from-scratch PPO implementation** with a hand-written reward function instead of DPO.

The reward function for brevity:

World!" gets -4.0 from exclamation marks)

Implement a `brevity_reward(response: str) -> float` function that computes the total reward. Then plug this into your PPO training loop from Part 2 in place of the neural reward model.


> **Important:** Log the **per-sample reward values for every batch** during RLVR training. You will need these logs to answer Theoretical Question 4 later. If you do not log them now, you will have to retrain.

````python
import re
 
FILLER_PHRASES = [
    "Certainly", "Of course", "Sure", "Absolutely", "Great question",
    "Good question", "That's a great question", "Let me", "I'd be happy to",
    "I would be happy to", "I'm happy to", "Happy to help", "Glad you asked",
    "That's an interesting", "That's a really good", "Thanks for asking",
    "Thank you for", "No problem", "Actually,", "Basically,", "Essentially,",
    "In fact,", "To be honest", "I think", "I believe", "I hope this helps",
    "Hope that helps", "Does that make sense", "Let me know if", "Feel free to",
    "Don't hesitate",
]
SENTENCE_START_FILLERS = ["Well,", "So,"]
 
 
def brevity_reward(response):
    """Compute rule-based brevity reward for ZenBot-RLVR."""
    # -0.1 per word
    words  = response.split()
    reward = -0.1 * len(words)
 
    # -2.0 per filler phrase occurrence
    for phrase in FILLER_PHRASES:
        count   = response.count(phrase)
        reward -= 2.0 * count
 
    for phrase in SENTENCE_START_FILLERS:
        pattern = r'(?:^|(?<=[.!?]\s))' + re.escape(phrase)
        count   = len(re.findall(pattern, response))
        reward -= 2.0 * count
 
    # -2.0 per exclamation mark
    exclaim_count = response.count("!")
    reward       -= 2.0 * exclaim_count
 
    return reward
 
 
def train_rlvr_zenbot(policy, ref_policy, value_head, tokenizer, prompts,
                      num_steps=200, beta=0.1):
    """Train ZenBot with PPO + rule-based brevity reward (RLVR)."""
    # Reuse your PPO implementation from Part 2
    # Replace RM scoring with brevity_reward()
    device = next(policy.parameters()).device
    policy.to(device)
    ref_policy.to(device)
    value_head.to(device)
 
    policy_optimizer = AdamW(
        filter(lambda p: p.requires_grad, policy.parameters()), lr=1e-5
    )
    value_optimizer  = AdamW(value_head.parameters(), lr=1e-4)
 
    metrics = {"mean_reward": [], "rewards": []}
 
    for step in range(1, num_steps + 1):
        idxs  = torch.randperm(len(prompts))[:4].tolist()
        batch = [prompts[i] for i in idxs]
 
        policy.eval()
        all_rewards = []
 
        for prompt in batch:
            formatted = f"### Prompt:\n{prompt}\n\n### Response:\n"
            enc       = tokenizer(formatted, return_tensors="pt").to(device)
 
            with torch.no_grad():
                out = policy.generate(
                    **enc,
                    max_new_tokens=80,
                    do_sample=True,
                    temperature=0.9,
                    pad_token_id=tokenizer.eos_token_id,
                )
 
            response_ids  = out[0][enc["input_ids"].shape[1]:]
            response_text = tokenizer.decode(response_ids, skip_special_tokens=True)
            r             = brevity_reward(response_text)
            all_rewards.append(r)
 
        mean_reward = sum(all_rewards) / len(all_rewards)
        metrics["mean_reward"].append(mean_reward)
        metrics["rewards"].extend(all_rewards)
 
        if step % 20 == 0:
            print(f"[RLVR] Step {step}/{num_steps} | Mean Reward: {mean_reward:.3f}")
 
    return metrics
 
````

## Part 5: GRPO (100 marks)

Group Relative Policy Optimization (GRPO) is a recent technique from [DeepSeek-R1](https://arxiv.org/abs/2501.12948). Unlike PPO, GRPO does not need a separate value (critic) network. Instead, for each prompt it generates a group of responses, scores all of them, and uses the relative ranking within the group to compute advantages.

The idea is simple: generate $G$ responses for the same prompt, compute rewards for each, normalize the rewards within the group (subtract mean, divide by std), and use these normalized rewards as advantages in a policy gradient update.

The GRPO objective:
$$ J_{\text{GRPO}}(\theta) = \mathbb{E}_{x \sim D} \left[ \frac{1}{G} \sum_{i=1}^{G} \min\left( \frac{\pi_\theta(y_i|x)}{\pi_{\text{old}}(y_i|x)} \hat{A}_i, \; \text{clip}\left(\frac{\pi_\theta(y_i|x)}{\pi_{\text{old}}(y_i|x)}, 1-\epsilon, 1+\epsilon\right) \hat{A}_i \right) \right] - \beta \, \text{KL}[\pi_\theta \| \pi_{\text{ref}}] $$

where $\hat{A}_i = \frac{r_i - \text{mean}(\mathbf{r})}{\text{std}(\mathbf{r})}$ and $\mathbf{r} = [r_1, \ldots, r_G]$ are the rewards for the $G$ responses.

No critic. No value head. Just generate, score, rank, update.

### Why GRPO here?

You have already seen PPO (needs 4 models in VRAM), DPO (offline, no generation), and RLVR (PPO with rule-based reward). GRPO sits in the middle ground. It is online like PPO (generates during training), but lighter on memory since there is no critic network. This makes it especially relevant for aligning small models on limited hardware.

### Your task

**Implement GRPO from scratch.** Train a **SafeBot-GRPO** variant using the same SafeBot reward model from Part 1. Do this for both GPT-2 and Gemma 3. Compare SafeBot-GRPO with your SafeBot-PPO from Part 2.

You need to implement:
1. For each prompt, generate $G$ responses (try $G=4$)
2. Score all $G$ responses with the SafeBot reward model
3. Normalize rewards within the group to get advantages $\hat{A}_i$
4. Compute the clipped policy gradient loss (same clipping as PPO, but using group-normalized advantages)
5. Add a KL penalty term against $\pi_{\text{ref}}$
6. Update the policy

````python
def grpo_step(policy, ref_policy, rm, tokenizer, prompts,
              group_size=4, beta=0.1, clip_eps=0.2):
    """One GRPO update step (no value network)."""
    device = next(policy.parameters()).device
    policy.train()
 
    total_loss  = 0.0
    all_rewards = []
 
    for prompt in prompts:
        formatted = f"### Prompt:\n{prompt}\n\n### Response:\n"
        enc       = tokenizer(formatted, return_tensors="pt").to(device)
        prompt_len = enc["input_ids"].shape[1]
 
        # For each prompt, generate group_size responses
        policy.eval()
        responses_ids  = []
        responses_text = []
        with torch.no_grad():
            for _ in range(group_size):
                out = policy.generate(
                    **enc,
                    max_new_tokens=100,
                    do_sample=True,
                    temperature=0.9,
                    pad_token_id=tokenizer.eos_token_id,
                )
                responses_ids.append(out[0])
                text = tokenizer.decode(out[0][prompt_len:], skip_special_tokens=True)
                responses_text.append(text)
 
        # Score all with RM
        rewards = torch.tensor(
            [score_with_rm(rm, tokenizer, prompt, r) for r in responses_text],
            dtype=torch.float32, device=device,
        )
        all_rewards.extend(rewards.tolist())
 
        # Normalize rewards within each group (subtract mean, divide by std)
        r_mean = rewards.mean()
        r_std  = rewards.std().clamp(min=1e-8)
        # Use normalized rewards as advantages
        advantages = (rewards - r_mean) / r_std     # (G,)
 
        policy.train()
        group_loss = 0.0
 
        for g_idx, (gen_ids, adv) in enumerate(zip(responses_ids, advantages)):
            full_ids  = gen_ids.unsqueeze(0)
            full_mask = (full_ids != tokenizer.pad_token_id).long()
 
            # PPO-style clipped loss (no value loss needed)
            outputs_new = policy(input_ids=full_ids, attention_mask=full_mask)
            lp_new = F.log_softmax(outputs_new.logits[:, :-1, :], dim=-1)
            target_ids = full_ids[:, 1:]
            per_tok_new = lp_new.gather(2, target_ids.unsqueeze(-1)).squeeze(-1)  # (1,T-1)
            response_lp_new = per_tok_new[:, prompt_len - 1:].sum(dim=-1)          # (1,)
 
            with torch.no_grad():
                outputs_old = policy(input_ids=full_ids, attention_mask=full_mask)
                lp_old      = F.log_softmax(outputs_old.logits[:, :-1, :], dim=-1)
                per_tok_old = lp_old.gather(2, target_ids.unsqueeze(-1)).squeeze(-1)
                response_lp_old = per_tok_old[:, prompt_len - 1:].sum(dim=-1)
 
                outputs_ref = ref_policy(input_ids=full_ids, attention_mask=full_mask)
                lp_ref      = F.log_softmax(outputs_ref.logits[:, :-1, :], dim=-1)
                per_tok_ref = lp_ref.gather(2, target_ids.unsqueeze(-1)).squeeze(-1)
                response_lp_ref = per_tok_ref[:, prompt_len - 1:].sum(dim=-1)
 
            ratio = torch.exp(response_lp_new - response_lp_old)
            surr1 = ratio * adv
            surr2 = torch.clamp(ratio, 1 - clip_eps, 1 + clip_eps) * adv
            pg_loss = -torch.min(surr1, surr2)
 
            kl_penalty = beta * (response_lp_new - response_lp_ref).mean()
            group_loss += (pg_loss + kl_penalty).squeeze()
 
        group_loss = group_loss / group_size
        group_loss.backward()
        total_loss += group_loss.item()
 
    return {
        "mean_reward": sum(all_rewards) / len(all_rewards),
        "loss":        total_loss / len(prompts),
        "rewards":     all_rewards,
    }
 
 
def train_grpo(policy, ref_policy, rm, tokenizer, prompts,
               num_steps=200, group_size=4, beta=0.1):
    """Full GRPO training loop for SafeBot. Track reward, KL per step."""
    device = next(policy.parameters()).device
    policy.to(device)
    ref_policy.to(device)
    for p in ref_policy.parameters():
        p.requires_grad_(False)
 
    optimizer = AdamW(
        filter(lambda p: p.requires_grad, policy.parameters()), lr=1e-5
    )
 
    metrics = {"mean_reward": [], "loss": [], "rewards": []}
 
    for step in range(1, num_steps + 1):
        idxs  = torch.randperm(len(prompts))[:4].tolist()
        batch = [prompts[i] for i in idxs]
 
        optimizer.zero_grad()
        step_metrics = grpo_step(
            policy, ref_policy, rm, tokenizer, batch,
            group_size=group_size, beta=beta,
        )
        torch.nn.utils.clip_grad_norm_(policy.parameters(), 1.0)
        optimizer.step()
 
        metrics["mean_reward"].append(step_metrics["mean_reward"])
        metrics["loss"].append(step_metrics["loss"])
        metrics["rewards"].extend(step_metrics["rewards"])
 
        if step % 20 == 0:
            print(f"[GRPO] Step {step}/{num_steps} | Reward: {step_metrics['mean_reward']:.3f}")
 
    return metrics
 
````
