````python
class ReplayBuffer():
    def __init__(self, buffer_size, batch_size, seed):
        # this function creates the relevant data-structures, and intializes all relevant variables
        #Your code goes in here
        self.memory = deque(maxlen=buffer_size)
        self.batch_size = batch_size
        random.seed(seed)

        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def store(self, state, action, reward, next_state, done):
        #stores the experiences, based on parameters in init
        #
        #this function does not return anything
        #
        #Your code goes in here
        experience = self.experience(state, action, reward, next_state, done)
        self.memory.append(experience)

        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def sample(self, batchSize):
        # this method returns batchSize number of experiences
        # this function returns experiences samples
        #
        #Your code goes in here
        experiencesList = random.sample(self.memory, min(batchSize, len(self.memory)))


        return experiencesList
````

````python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import gymnasium as gym
import time

# def __init__


class REINFORCE():
    def __init__(self, env, seed, gamma,
                 optimizerLR,
                 MAX_TRAIN_EPISODES, MAX_EVAL_EPISODES):

        self.env = env
        self.gamma = gamma
        self.MAX_TRAIN_EPISODES = MAX_TRAIN_EPISODES
        self.MAX_EVAL_EPISODES  = MAX_EVAL_EPISODES

        torch.manual_seed(seed)
        np.random.seed(seed)
        self.env.reset(seed=seed)

        inDim  = env.observation_space.shape[0]   # 3 for Pendulum-v1
        outDim = env.action_space.shape[0]         # 1 for Pendulum-v1

       
        self.policy = nn.Sequential(
            nn.Linear(inDim, 256),
            nn.ReLU(),
            nn.Linear(256, 256),
            nn.ReLU(),
            nn.Linear(256, outDim)
        )

        
        self.log_std = nn.Parameter(torch.full((outDim,), -0.5))
        self.policy.register_parameter('log_std', self.log_std)

        self.optimizer = optim.Adam(self.policy.parameters(), lr=optimizerLR)

        self.scheduler = optim.lr_scheduler.StepLR(
            self.optimizer, step_size=200, gamma=0.5
        )

        self.start_time = time.time()
        self.initBookKeeping()
````

````python
# def initBookKeeping
class REINFORCE():

    def initBookKeeping(self):
        self.train_rewards = []
        self.eval_rewards  = []
        self.steps         = []
        self.train_times   = []
        self.wall_times    = []
````

````python
# def performBookKeeping
class REINFORCE():
    def performBookKeeping(self, ep_reward, step, elapsed):
        self.train_rewards.append(ep_reward)
        self.steps.append(step)
        self.train_times.append(elapsed)
        self.wall_times.append(time.time() - self.start_time)
````

````python
# def runREINFORCE
class REINFORCE():
    def runREINFORCE(self):
        start_time = time.time()

        for ep in range(self.MAX_TRAIN_EPISODES):
            self.trainAgent(ep, start_time)

            if (ep + 1) % 10 == 0:
                eval_r = self.evaluateAgent()
                self.eval_rewards.append((ep + 1, eval_r))

            
            self.scheduler.step()

        return (self.train_rewards,
                self.eval_rewards,
                self.steps,
                self.train_times,
                self.wall_times)
````

````python
# def trainAgent
class REINFORCE():
        def trainAgent(self, ep, start_time):
        state, _ = self.env.reset()
        done, truncated = False, False

        log_probs = []
        rewards   = []
        ep_reward = 0
        step      = 0

        while not (done or truncated):
            state_t = torch.tensor(state, dtype=torch.float32)

            mean = self.policy(state_t)
            std  = self.log_std.exp().expand_as(mean)
            dist = torch.distributions.Normal(mean, std)

            raw_action = dist.sample()

            # tanh correction (unchanged)
            log_prob = (dist.log_prob(raw_action)
                        - torch.log(1 - torch.tanh(raw_action)**2 + 1e-6)
                       ).sum()

            action        = torch.tanh(raw_action)
            action_scaled = action * 2.0

            state, reward, done, truncated, _ = self.env.step(
                action_scaled.detach().numpy()
            )

            log_probs.append(log_prob)
            rewards.append(reward)
            ep_reward += reward
            step      += 1

        self.trainPolicyNetwork(log_probs, rewards)

        elapsed = time.time() - start_time
        self.performBookKeeping(ep_reward, step, elapsed)
````

````python
# def trainPolicyNetwork

class REINFORCE():
        def trainPolicyNetwork(self, log_probs, rewards):
        returns = []
        R = 0

        for r in reversed(rewards):
            R = r + self.gamma * R
            returns.insert(0, R)

        returns = torch.tensor(returns, dtype=torch.float32)

        # CHANGE 3: subtract mean return as baseline BEFORE normalising
        # reduces gradient variance without introducing a learned network
        returns = returns - returns.mean()
        returns = returns / (returns.std() + 1e-8)

        loss = 0
        for log_prob, R in zip(log_probs, returns):
            loss += -log_prob * R

        self.optimizer.zero_grad()
        loss.backward()

        # CHANGE 4: gradient clipping — prevents large gradient spikes
        torch.nn.utils.clip_grad_norm_(
            self.policy.parameters(), max_norm=0.5
        )

        self.optimizer.step()
````

````python
# def evaluateAgent
class REINFORCE():
        def evaluateAgent(self):
        total_reward = 0

        for _ in range(self.MAX_EVAL_EPISODES):
            state, _ = self.env.reset()
            done, truncated = False, False
            ep_reward = 0

            while not (done or truncated):
                state_t = torch.tensor(state, dtype=torch.float32)

                mean   = self.policy(state_t)
                action = torch.tanh(mean) * 2.0

                state, reward, done, truncated, _ = self.env.step(
                    action.detach().numpy()
                )
                ep_reward += reward

            total_reward += ep_reward

        return total_reward / self.MAX_EVAL_EPISODES


# ── Usage ─────────────────────────────────────────────────────
if __name__ == "__main__":
    env = gym.make("Pendulum-v1")

    agent = REINFORCE(
        env                = env,
        seed               = 42,
        gamma              = 0.99,
        optimizerLR        = 1e-3,       
        MAX_TRAIN_EPISODES = 1000,        
        MAX_EVAL_EPISODES  = 10,
    )

    train_r, eval_r, steps, t_times, w_times = agent.runREINFORCE()
    print(f"Final eval reward: {eval_r[-1][1]:.2f}")
````

````python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import gymnasium as gym
import time
from itertools import count
import matplotlib.pyplot as plt


class PolicyNetwork(nn.Module):
    def __init__(self, inDim, outDim, hidden=256):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(inDim, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, outDim)
        )
        self.log_std = nn.Parameter(torch.full((outDim,), -0.5))

    def forward(self, x):
        return self.net(x)

    def selectAction(self, s):
        s_t   = torch.tensor(s, dtype=torch.float32)
        mean  = self.net(s_t)
        std   = self.log_std.exp().expand_as(mean)
        dist  = torch.distributions.Normal(mean, std)

        raw        = dist.rsample()
        log_prob   = (dist.log_prob(raw)
                      - torch.log(1 - torch.tanh(raw)**2 + 1e-6)).sum()
        entropy    = dist.entropy().sum()
        action     = torch.tanh(raw) * 2.0          # scale to [-2, 2]
        return action.detach().numpy(), log_prob, entropy

    def selectGreedyAction(self, s):
        s_t    = torch.tensor(s, dtype=torch.float32)
        mean   = self.net(s_t)
        action = torch.tanh(mean) * 2.0
        return action.detach().numpy()


class ValueNetwork(nn.Module):
    def __init__(self, inDim, hidden=256):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(inDim, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, 1)
        )

    def forward(self, x):
        return self.net(x)


def createPolicyNetwork(inDim, outDim):
    return PolicyNetwork(inDim, outDim)

def createValueNetwork(inDim, outDim, hidden=256):
    
    return ValueNetwork(inDim, hidden)

````

````python
# def trainAgent
class VPG():
    def trainAgent(self):
        results    = []
        epoch_start = time.time()

        for e in range(self.MAX_TRAIN_EPISODES):
            s, _  = self.env.reset()
            rewards, logProbs, entropies, values = [], [], [], []
            totalReward = 0
            ep_steps    = 0

            
            done, truncated = False, False
            while not (done or truncated):
                a, logp_a, entropy = self.pNetwork.selectAction(s)
                v = self.vNetwork(
                    torch.tensor(s, dtype=torch.float32))

                s, r, done, truncated, _ = self.env.step(a)

                rewards.append(r)
                logProbs.append(logp_a)
                entropies.append(entropy)
                values.append(v.squeeze())

                totalReward += r
                ep_steps    += 1

            self.trainPolicyNetwork((rewards, logProbs, entropies, values))

            elapsed = time.time() - epoch_start
            self.performBookKeeping(totalReward, ep_steps, elapsed)

            
            if (e + 1) % 10 == 0:
                eval_mean, _ = self.evaluateAgent()
                self.evalRewards.append((e + 1, eval_mean))

            results.append((e, totalReward))

        return results
````

````python
# def trainPolicyNetwork
class VPG():
    def trainPolicyNetwork(self, experiences):
        rewards, logProbs, entropies, values = experiences

        returns = []
        G = 0
        for r in reversed(rewards):
            G = r + self.gamma * G
            returns.insert(0, G)

        returns     = torch.tensor(returns, dtype=torch.float32)
        values_t    = torch.stack(values)          
        logProbs_t  = torch.stack(logProbs)
        entropies_t = torch.stack(entropies)

        
        returns = (returns - returns.mean()) / (returns.std() + 1e-8)

        
        deltas = returns - values_t.detach()

        # ── policy loss ───────────────────────────────────────────
        policyLoss      = -(deltas * logProbs_t).mean()
        entropyLoss     = -entropies_t.mean()
        totalPolicyLoss = policyLoss + self.beta * entropyLoss

        self.policyOptimizerFn.zero_grad()
        totalPolicyLoss.backward()
        torch.nn.utils.clip_grad_norm_(self.pNetwork.parameters(), 1.0)
        self.policyOptimizerFn.step()

       
        valueLoss = 0.5 * ((returns - values_t) ** 2).mean()

        self.valueOptimizerFn.zero_grad()
        valueLoss.backward()
        torch.nn.utils.clip_grad_norm_(self.vNetwork.parameters(), 1.0)
        self.valueOptimizerFn.step()
````

````python
# def evaluateAgent
class VPG():
    def evaluateAgent(self, greedy=True):
        rewards = []

        for _ in range(self.MAX_EVAL_EPISODES):
            rs   = 0
            s, _ = self.env.reset()

            for _ in count():
                if greedy:
                    a = self.pNetwork.selectGreedyAction(s)
                else:
                    a, _, _ = self.pNetwork.selectAction(s)

                s, r, done, truncated, _ = self.env.step(a)
                rs += r

                if done or truncated:
                    rewards.append(rs)
                    break

        mean_r = np.mean(rewards)
        std_r  = np.std(rewards)
        return mean_r, std_r


````

# Asynchronous Advantange Actor-Critic (A3C) [50 marks]
<a id="a3c"></a>

A3C (Mnih et al., 2016) was one of the first successful deep RL algorithms to train on CPU using parallel workers. Each worker maintains its own copy of the environment and policy, collects experience independently, and asynchronously pushes gradient updates to a shared global network.

**Key ideas:**
- Multiple worker processes each interact with their own environment copy.
- Each worker computes gradients locally and applies them to a **shared global network** asynchronously (no waiting for other workers).
- The asynchronous updates provide natural exploration diversity (different workers explore different parts of the state space).
- Uses n-step returns and advantage estimation similar to A2C.

**Update rule per worker:**
$$\theta \leftarrow \theta + \alpha \nabla_\theta \log \pi_\theta(a_t|s_t) \cdot A(s_t, a_t)$$

where the gradient is computed by the worker and applied to the shared $\theta$.

**Note:** True A3C requires `multiprocessing` with shared memory. Below we implement it using `torch.multiprocessing` with shared parameters.

```
class SharedAdam(optim.Adam)   # Adam optimizer with shared state
class A3CAgent(nn.Module)      # Same architecture as A2C
def a3c_worker(...)            # Worker process
def train_a3c(...)             # Main launcher
```

````python
# ======================================================================
# A3C Implementation
# ======================================================================

import torch.multiprocessing as mp

# set start method once at top
try:
    mp.set_start_method('spawn')
except RuntimeError:
    pass

# A3C Hyperparameters
a3c_gym_id                  = "Hopper-v4"
a3c_learning_rate           = 1e-3
a3c_gamma                   = 0.99
a3c_num_workers             = 4
a3c_num_steps               = 5
a3c_max_episodes_per_worker = 120
a3c_ent_coef                = 0.01
a3c_vf_coef                 = 0.5
a3c_max_grad_norm           = 0.5
````

````python
# ======================================================================
# SharedAdam: Adam optimizer with shared state across processes
# ======================================================================

class SharedAdam(optim.Adam):

    def __init__(self, params, lr=1e-3, betas=(0.9, 0.999), eps=1e-8):
        super().__init__(params, lr=lr, betas=betas, eps=eps)
        for group in self.param_groups:
            for p in group['params']:
                if p.requires_grad:
                    state                  = self.state[p]
                    state['step']          = torch.zeros(1, dtype=torch.int64).share_memory_()
                    state['exp_avg']       = torch.zeros_like(p.data).share_memory_()
                    state['exp_avg_sq']    = torch.zeros_like(p.data).share_memory_()
````

````python
# ======================================================================
# A3C Agent
# ======================================================================

class A3CAgent(nn.Module):
    """
    A3C agent with separate actor and critic networks.
    Supports both discrete and continuous action spaces.
    """

    def __init__(self, obs_dim, act_dim, is_discrete=False):
        """
        Initialize actor and critic networks.
        obs_dim     : Dimension of observation space
        act_dim     : Dimension of action space
        is_discrete : Whether the action space is discrete
        """
        super().__init__()
        self.is_discrete = is_discrete

        self.actor = nn.Sequential(
            nn.Linear(obs_dim, 128),
            nn.ReLU(),
            nn.Linear(128, act_dim)
        )

        self.critic = nn.Sequential(
            nn.Linear(obs_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 1)
        )

        # learnable std for continuous actions
        if not is_discrete:
            self.log_std = nn.Parameter(torch.zeros(act_dim))

    def get_value(self, x):
        """
        Return the value estimate for a given state using the critic.
        """
        return self.critic(x)

    def get_action_and_value(self, x, action=None):
        """
        Compute action, log probability, entropy, and value.

        For discrete actions:
            create a categorical distribution from actor logits.

        For continuous actions:
            create a normal distribution using mean and std.

        If action is None, sample from the distribution.
        Return action, log probability, entropy, and value.
        """
        x      = torch.tensor(x, dtype=torch.float32)
        logits = self.actor(x)

        if self.is_discrete:
            dist = torch.distributions.Categorical(logits=logits)
        else:
            mean = logits
            # clamp log_std to prevent NaN
            std  = self.log_std.clamp(-2, 2).exp().expand_as(mean)
            dist = torch.distributions.Normal(mean, std)

        if action is None:
            action = dist.sample()

        if self.is_discrete:
            logprob = dist.log_prob(action)
            entropy = dist.entropy()
        else:
            logprob = dist.log_prob(action).sum(-1)
            entropy = dist.entropy().sum(-1)
            # fix: removed torch.tanh — Hopper actions already bounded

        value = self.critic(x).squeeze(-1)

        return action, logprob, entropy, value
````

````python
# ======================================================================
# A3C Worker
# ======================================================================

def a3c_worker(worker_id, global_agent, global_optimizer, grad_lock,
               env_id, seed, num_episodes,
               gamma, num_steps, ent_coef, vf_coef, max_grad_norm,
               result_queue):
    """
    A3C worker — runs as a separate OS process.

    Each worker independently:
      1. Syncs local weights from the shared global network.
      2. Collects a rollout of num_steps transitions.
      3. Computes n-step returns and advantage.
      4. Computes gradients on its LOCAL network copy.
      5. Pushes those gradients to the GLOBAL network (under lock).
      6. Sends episode reward to main process via result_queue.

    The grad_lock serialises gradient push + optimizer.step() to prevent
    two workers from corrupting each other's gradients.
    """
    env         = gym.make(env_id)
    obs_dim     = env.observation_space.shape[0]
    is_discrete = hasattr(env.action_space, "n")
    act_dim     = env.action_space.n if is_discrete \
                  else env.action_space.shape[0]

    local_agent = A3CAgent(obs_dim, act_dim, is_discrete)
    local_agent.load_state_dict(global_agent.state_dict())

    for ep in range(num_episodes):
        s, _      = env.reset(seed=seed + worker_id)
        done      = False
        truncated = False
        ep_reward = 0

        rewards, logprobs, entropies, values = [], [], [], []

        while not (done or truncated):
            s_tensor = torch.tensor(s, dtype=torch.float32).unsqueeze(0)
            a, logp, ent, v = local_agent.get_action_and_value(s_tensor)

            action = a.item() if is_discrete \
                     else a.squeeze(0).detach().numpy()

            s, r, done, truncated, _ = env.step(action)

            rewards.append(r)
            logprobs.append(logp.squeeze())
            entropies.append(ent.squeeze())
            values.append(v.squeeze())
            ep_reward += r

            if len(rewards) >= num_steps or done or truncated:

                # bootstrap value
                with torch.no_grad():
                    if done:
                        R = 0
                    else:
                        s_tensor = torch.tensor(
                            s, dtype=torch.float32).unsqueeze(0)
                        R = local_agent.get_value(s_tensor).item()

                # compute returns
                returns = []
                for rwd in reversed(rewards):
                    R = rwd + gamma * R
                    returns.insert(0, R)

                returns_t   = torch.tensor(returns, dtype=torch.float32)
                values_t    = torch.stack(values)
                logprobs_t  = torch.stack(logprobs)
                entropies_t = torch.stack(entropies)

                advantages = returns_t - values_t.detach()

                # normalize advantage for stability
                if len(advantages) > 1:
                    advantages = (advantages - advantages.mean()) / \
                                 (advantages.std() + 1e-8)

                policy_loss = -(logprobs_t * advantages).mean() \
                              - ent_coef * entropies_t.mean()
                value_loss  = 0.5 * (returns_t - values_t).pow(2).mean()
                loss        = policy_loss + vf_coef * value_loss

                # backprop on LOCAL network
                local_agent.zero_grad()
                loss.backward()
                torch.nn.utils.clip_grad_norm_(
                    local_agent.parameters(), max_grad_norm)

                # push gradients to GLOBAL network under lock
                with grad_lock:
                    global_optimizer.zero_grad()
                    for lp, gp in zip(local_agent.parameters(),
                                      global_agent.parameters()):
                        gp._grad = lp.grad.detach()
                    global_optimizer.step()

                # sync local with global
                local_agent.load_state_dict(global_agent.state_dict())

                rewards, logprobs, entropies, values = [], [], [], []

        result_queue.put(ep_reward)

    result_queue.put(None)
    env.close()
````

````python
# ======================================================================
# A3C Launcher
# ======================================================================

def train_a3c(env_id=a3c_gym_id, seed=1, num_workers=a3c_num_workers,
              episodes_per_worker=a3c_max_episodes_per_worker,
              gamma=a3c_gamma, num_steps=a3c_num_steps,
              lr=a3c_learning_rate,
              ent_coef=a3c_ent_coef, vf_coef=a3c_vf_coef,
              max_grad_norm=a3c_max_grad_norm):
    """
    Launches A3C training with truly parallel worker processes.

    1. Creates global_agent in shared memory.
    2. Creates SharedAdam (optimizer state also in shared memory).
    3. Creates mp.Lock for gradient push serialisation.
    4. Spawns num_workers mp.Process instances (all start simultaneously).
    5. Main process collects rewards from Queue while workers train.
    6. p.join() waits for all workers to finish.
    """
    env         = gym.make(env_id)
    obs_dim     = env.observation_space.shape[0]
    is_discrete = hasattr(env.action_space, "n")
    act_dim     = env.action_space.n if is_discrete \
                  else env.action_space.shape[0]
    env.close()

    # 1. create global_agent in shared memory
    global_agent = A3CAgent(obs_dim, act_dim, is_discrete)
    global_agent.share_memory()
    global_agent.train()

    # 2. create SharedAdam
    global_optimizer = SharedAdam(global_agent.parameters(), lr=lr)

    # 3. create mp.Lock
    grad_lock    = mp.Lock()
    result_queue = mp.Queue()

    # 4. spawn workers
    workers = []
    for i in range(num_workers):
        p = mp.Process(
            target = a3c_worker,
            args   = (i, global_agent, global_optimizer, grad_lock,
                      env_id, seed, episodes_per_worker,
                      gamma, num_steps, ent_coef, vf_coef,
                      max_grad_norm, result_queue)
        )
        p.start()
        workers.append(p)

    # 5. collect rewards from queue
    rewards          = []
    finished_workers = 0

    while finished_workers < num_workers:
        r = result_queue.get()
        if r is None:
            finished_workers += 1
        else:
            rewards.append(r)

    # 6. wait for all workers
    for p in workers:
        p.join()

    return rewards, global_agent
````

````python
rewards, _ = train_a3c()

rewards = np.array(rewards)

def moving_average(x, window=20):
    if len(x) < window:
        return x
    return np.convolve(x, np.ones(window)/window, mode='valid')

plt.figure(figsize=(10,5))

plt.plot(rewards, alpha=0.3, label="Raw Reward")

smoothed = moving_average(rewards, 20)
plt.plot(np.arange(len(smoothed)) + 19, smoothed, linewidth=2, label="Smoothed")

plt.title("A3C on Hopper-v4")
plt.xlabel("Episode")
plt.ylabel("Return")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
````

**Answer 1 :** In A3C, multiple agents work independently and update the main model whenever they’re ready (not waiting for others). It helps in faster learning and more diverse experiences as each agent sees different things. It also helps avoiding getting stuck in bad strategies. On the other hand, synchronous learning waits for all the agents to fininsh and then update the main model, hence it is slower and less diverse.

**Answer 2 :** In A3C, all the agents compute the update i.e. gradient independently, so, before they apply the update to the main model, other agents have already updated the model. Now, the update is based on old information but the model has already changed. This is called the stale gradient. It results that the update may not be perfectly correct anymore and can make learning a bit noisy or unstable.

**Answer 3 :** DQN stores lots of past experiences in a replay buffer, so it requires high memory. On the other hand, A3C does not store past data, so it requires much lower memory. Talking about the training, DQN is off policy and learns from past experiences, it is more stable but slow, while the A3C is on policy, it learns live while interacting , hence it is faster and more dynamic.

# Advantange Actor-Critic (A2C) [50 marks]
<a id="a2c"></a>

A2C is the synchronous, deterministic variant of A3C. Instead of using asynchronous workers, A2C waits for all parallel workers to finish their rollouts before performing a single gradient update. This makes it simpler to implement and often equally effective.

**Key ideas:**
- Uses an Actor-Critic architecture: the actor outputs a policy $\pi(a|s)$ and the critic estimates $V(s)$.
- The advantage $A(s,a) = R_t - V(s_t)$ (or using GAE) is used to reduce variance in the policy gradient.
- Multiple environments run in parallel (vectorized) and all updates are synchronous.
- The loss combines: policy gradient loss, value loss, and an entropy bonus for exploration.

**Policy Gradient with Advantage:**
$$\nabla_\theta J(\theta) = \mathbb{E}\left[\nabla_\theta \log \pi_\theta(a_t|s_t) \cdot A(s_t, a_t)\right]$$

````python
class A2CAgent(nn.Module):
    def __init__(self, envs):
        super().__init__()

        obs_dim = envs.observation_space.shape[0]
        self.is_discrete = hasattr(envs.action_space, "n")

        if self.is_discrete:
            act_dim = envs.action_space.n
        else:
            act_dim = envs.action_space.shape[0]

        def init(layer, std=np.sqrt(2)):
            nn.init.orthogonal_(layer.weight, std)
            nn.init.constant_(layer.bias, 0.0)
            return layer

        # Critic
        self.critic = nn.Sequential(
            init(nn.Linear(obs_dim, 64)), nn.Tanh(),
            init(nn.Linear(64, 64)), nn.Tanh(),
            init(nn.Linear(64, 1), std=1.0),
        )

        # Actor
        self.actor = nn.Sequential(
            init(nn.Linear(obs_dim, 64)), nn.Tanh(),
            init(nn.Linear(64, 64)), nn.Tanh(),
            init(nn.Linear(64, act_dim), std=0.01),
        )

        if not self.is_discrete:
            self.log_std = nn.Parameter(torch.zeros(act_dim))

    def get_value(self, x):
        if not isinstance(x, torch.Tensor):
            x = torch.tensor(x, dtype=torch.float32)
        return self.critic(x).squeeze(-1)

    def get_action_and_value(self, x, action=None):
        if not isinstance(x, torch.Tensor):
            x = torch.tensor(x, dtype=torch.float32)

        logits = self.actor(x)
        value = self.get_value(x)

        if self.is_discrete:
            dist = torch.distributions.Categorical(logits=logits)
        else:
            std = self.log_std.exp().expand_as(logits)
            dist = torch.distributions.Normal(logits, std)

        if action is None:
            action = dist.sample()

        if self.is_discrete:
            log_prob = dist.log_prob(action)
            entropy = dist.entropy()
        else:
            log_prob = dist.log_prob(action).sum(-1)
            entropy = dist.entropy().sum(-1)

        return action, log_prob, entropy, value
````

````python
def a2c_worker(worker_id, global_agent,
               env_id, seed, num_steps, num_updates,
               shared_obs, shared_actions, shared_logprobs,
               shared_rewards, shared_dones, shared_values,
               shared_next_obs, shared_next_done,
               collect_barrier, update_barrier, ep_queue):

    import gymnasium as gym
    import torch
    import numpy as np

    env = gym.make(env_id)
    env.action_space.seed(seed + worker_id)
    torch.manual_seed(seed + worker_id)

    obs, _ = env.reset(seed=seed + worker_id)
    done = False
    ep_reward = 0.0

    for _ in range(num_updates):

        for step in range(num_steps):

            shared_obs[worker_id, step] = torch.tensor(obs, dtype=torch.float32)
            shared_dones[worker_id, step] = float(done)

            with torch.no_grad():
                action, logprob, _, value = global_agent.get_action_and_value(obs)

            # Store
            if global_agent.is_discrete:
                shared_actions[worker_id, step] = action
            else:
                shared_actions[worker_id, step] = action

            shared_logprobs[worker_id, step] = logprob
            shared_values[worker_id, step] = value

            # Step env
            if global_agent.is_discrete:
                act = action.item()
            else:
                act = action.numpy()

            next_obs, reward, terminated, truncated, _ = env.step(act)
            done = terminated or truncated

            shared_rewards[worker_id, step] = reward
            ep_reward += reward

            if done:
                ep_queue.put(ep_reward)
                ep_reward = 0.0
                next_obs, _ = env.reset()

            obs = next_obs

        shared_next_obs[worker_id] = torch.tensor(obs, dtype=torch.float32)
        shared_next_done[worker_id] = float(done)

        # Synchronize
        collect_barrier.wait()
        update_barrier.wait()

    env.close()
````

````python
a2c_gym_id = "Hopper-v4"
a2c_seed = 1
a2c_total_timesteps = 100_000
a2c_num_envs = 8
a2c_num_steps = 5
a2c_learning_rate = 3e-4
a2c_gamma = 0.99
a2c_gae_lambda = 0.95
a2c_use_gae = True
a2c_ent_coef = 0.01
a2c_vf_coef = 0.5
a2c_max_grad_norm = 0.5
a2c_norm_adv = True
````

````python
def train_a2c(env_id=a2c_gym_id, seed=a2c_seed, total_timesteps=a2c_total_timesteps,
              num_envs=a2c_num_envs, num_steps=a2c_num_steps, lr=a2c_learning_rate,
              gamma=a2c_gamma, gae_lambda=a2c_gae_lambda, use_gae=a2c_use_gae,
              ent_coef=a2c_ent_coef, vf_coef=a2c_vf_coef,
              max_grad_norm=a2c_max_grad_norm, norm_adv=a2c_norm_adv):

    import torch
    import torch.optim as optim
    import numpy as np
    import gymnasium as gym
    import torch.multiprocessing as mp

    torch.manual_seed(seed)
    np.random.seed(seed)

    tmp_env = gym.make(env_id)
    obs_dim = tmp_env.observation_space.shape[0]
    is_discrete = hasattr(tmp_env.action_space, "n")

    if is_discrete:
        act_dim = 1
    else:
        act_dim = tmp_env.action_space.shape[0]

    tmp_env.close()

    num_updates = total_timesteps // (num_envs * num_steps)

    # Proxy env
    class Proxy:
        pass

    proxy = Proxy()
    proxy.observation_space = type('', (), {'shape': (obs_dim,)})()
    if is_discrete:
        proxy.action_space = type('', (), {'n': act_dim})()
    else:
        proxy.action_space = type('', (), {'shape': (act_dim,)})()

    global_agent = A2CAgent(proxy)
    global_agent.is_discrete = is_discrete
    global_agent.share_memory()

    optimizer = optim.Adam(global_agent.parameters(), lr=lr)

    # Shared buffers
    shared_obs = torch.zeros(num_envs, num_steps, obs_dim).share_memory_()
    shared_actions = torch.zeros(num_envs, num_steps, act_dim).share_memory_()
    shared_logprobs = torch.zeros(num_envs, num_steps).share_memory_()
    shared_rewards = torch.zeros(num_envs, num_steps).share_memory_()
    shared_dones = torch.zeros(num_envs, num_steps).share_memory_()
    shared_values = torch.zeros(num_envs, num_steps).share_memory_()

    shared_next_obs = torch.zeros(num_envs, obs_dim).share_memory_()
    shared_next_done = torch.zeros(num_envs).share_memory_()

    collect_barrier = mp.Barrier(num_envs + 1)
    update_barrier = mp.Barrier(num_envs + 1)
    ep_queue = mp.Queue()

    # Start workers
    processes = []
    for wid in range(num_envs):
        p = mp.Process(target=a2c_worker, args=(
            wid, global_agent, env_id, seed,
            num_steps, num_updates,
            shared_obs, shared_actions, shared_logprobs,
            shared_rewards, shared_dones, shared_values,
            shared_next_obs, shared_next_done,
            collect_barrier, update_barrier, ep_queue
        ))
        p.start()
        processes.append(p)

    all_rewards = []

    for _ in range(num_updates):

        collect_barrier.wait()

        with torch.no_grad():
            next_value = global_agent.get_value(shared_next_obs)

        advantages = torch.zeros_like(shared_rewards)

        last_gae = 0
        for t in reversed(range(num_steps)):
            if t == num_steps - 1:
                next_nonterminal = 1.0 - shared_next_done
                next_values = next_value
            else:
                next_nonterminal = 1.0 - shared_dones[:, t + 1]
                next_values = shared_values[:, t + 1]

            delta = shared_rewards[:, t] + gamma * next_values * next_nonterminal - shared_values[:, t]
            last_gae = delta + gamma * gae_lambda * next_nonterminal * last_gae
            advantages[:, t] = last_gae

        returns = advantages + shared_values

        if norm_adv:
            advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

        b_obs = shared_obs.reshape(-1, obs_dim)
        b_actions = shared_actions.reshape(-1, act_dim)
        b_returns = returns.reshape(-1)
        b_advantages = advantages.reshape(-1)

        _, logprob, entropy, value = global_agent.get_action_and_value(b_obs, b_actions)

        policy_loss = -(b_advantages * logprob).mean()
        value_loss = vf_coef * (value - b_returns).pow(2).mean()
        entropy_loss = -ent_coef * entropy.mean()

        loss = policy_loss + value_loss + entropy_loss

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(global_agent.parameters(), max_grad_norm)
        optimizer.step()

        while not ep_queue.empty():
            all_rewards.append(ep_queue.get())

        update_barrier.wait()

    for p in processes:
        p.join()

    return all_rewards, global_agent
````

# Generalized Advantage Estimation (GAE) [50 marks]
<a id="gae"></a>

GAE (Schulman et al., 2016) provides a principled way to balance bias and variance in advantage estimation through a single parameter $\lambda \in [0, 1]$.

**The core idea:**

The 1-step advantage (TD error) is: $\delta_t^V = r_t + \gamma V(s_{t+1}) - V(s_t)$

The GAE advantage is an exponentially-weighted average of multi-step advantages:
$$\hat{A}_t^{\text{GAE}(\gamma, \lambda)} = \sum_{l=0}^{\infty} (\gamma \lambda)^l \delta_{t+l}^V$$

This can be computed recursively:
$$\hat{A}_t = \delta_t + \gamma \lambda \hat{A}_{t+1}$$

**Special cases:**
- $\lambda = 0$: GAE reduces to the 1-step TD advantage $\delta_t^V$ → **low variance, high bias**
- $\lambda = 1$: GAE reduces to the full Monte Carlo advantage → **high variance, low bias**
- $0 < \lambda < 1$: Smooth interpolation between the two extremes

Below we implement a standalone GAE module and demonstrate how different values of $\lambda$ affect learning performance.


````python
# ======================================================================
# GAE Implementation (standalone module)
# ======================================================================

def compute_gae(rewards, values, dones, next_value, gamma, gae_lambda):
    """
    Computes Generalized Advantage Estimation.

    Args:
        rewards    : tensor of shape (T, num_envs) - rewards at each timestep
        values     : tensor of shape (T, num_envs) - V(s_t) predictions
        dones      : tensor of shape (T, num_envs) - done flags
        next_value : tensor of shape (1, num_envs) - V(s_{T}) bootstrap value
        gamma      : discount factor
        gae_lambda : GAE lambda parameter (0=TD, 1=MC)

    Returns:
        advantages : tensor of shape (T, num_envs)
        returns    : tensor of shape (T, num_envs)
    """

    #Your code goes here
    T          = rewards.shape[0]
    advantages = torch.zeros_like(rewards)
    gae        = 0

    # FIX 1: flatten next_value to (num_envs,) for consistent shapes
    next_value = next_value.squeeze(0)

    for t in reversed(range(T)):
        if t == T - 1:
            # FIX 2: bootstrap uses next_value and next done = dones[t]
            # (the done at the last step tells us if episode ended)
            next_non_terminal = 1.0 - dones[t]
            next_vals         = next_value
        else:
            next_non_terminal = 1.0 - dones[t + 1]
            next_vals         = values[t + 1]

        delta  = (rewards[t]
                  + gamma * next_vals * next_non_terminal
                  - values[t])
        gae    = delta + gamma * gae_lambda * next_non_terminal * gae
        advantages[t] = gae

    returns = advantages + values
    return advantages, returns
````

````python

def moving_average(x, window=20):
    if len(x) < window:
        return x
    return np.convolve(x, np.ones(window) / window, mode='valid')



class A2CActor(nn.Module):
    def __init__(self, obs_dim, act_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(obs_dim, 256), nn.Tanh(),
            nn.Linear(256, 256),     nn.Tanh(),
            nn.Linear(256, act_dim)
        )
        self.log_std = nn.Parameter(torch.zeros(act_dim))

    def forward(self, x):
        mean = self.net(x)
        std  = self.log_std.clamp(-2, 2).exp().expand_as(mean)
        return torch.distributions.Normal(mean, std)

    def get_action(self, x):
        dist     = self.forward(x)
        action   = dist.sample()
        log_prob = dist.log_prob(action).sum(-1)
        entropy  = dist.entropy().sum(-1)
        return action, log_prob, entropy


class A2CCritic(nn.Module):
    def __init__(self, obs_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(obs_dim, 256), nn.Tanh(),
            nn.Linear(256, 256),     nn.Tanh(),
            nn.Linear(256, 1)
        )

    def forward(self, x):
        return self.net(x).squeeze(-1)



def train_a2c(env_id, total_timesteps, num_envs=4, num_steps=5,
              lr=3e-4, gamma=0.99, gae_lambda=0.95,
              ent_coef=0.01, vf_coef=0.5, max_grad_norm=0.5,
              seed=0):
    """
    Synchronous A2C with GAE.
    Returns list of episode rewards collected during training.
    """
    torch.manual_seed(seed)
    np.random.seed(seed)

    # vectorised envs
    envs    = gym.vector.SyncVectorEnv(
        [lambda i=i: gym.make(env_id) for i in range(num_envs)]
    )
    obs_dim = envs.single_observation_space.shape[0]
    act_dim = envs.single_action_space.shape[0]

    actor   = A2CActor(obs_dim, act_dim)
    critic  = A2CCritic(obs_dim)
    params  = list(actor.parameters()) + list(critic.parameters())
    optimizer = optim.Adam(params, lr=lr)

    obs, _       = envs.reset(seed=seed)
    obs          = torch.tensor(obs, dtype=torch.float32)

    episode_rewards = []
    ep_returns      = np.zeros(num_envs)
    total_steps     = 0

    while total_steps < total_timesteps:

        # ── collect num_steps rollout ──────────────────────────
        obs_buf      = torch.zeros(num_steps, num_envs, obs_dim)
        act_buf      = torch.zeros(num_steps, num_envs, act_dim)
        logp_buf     = torch.zeros(num_steps, num_envs)
        rew_buf      = torch.zeros(num_steps, num_envs)
        done_buf     = torch.zeros(num_steps, num_envs)
        val_buf      = torch.zeros(num_steps, num_envs)
        entropy_buf  = torch.zeros(num_steps, num_envs)

        for step in range(num_steps):
            with torch.no_grad():
                action, log_prob, entropy = actor.get_action(obs)
                value                    = critic(obs)

            next_obs, reward, terminated, truncated, infos = \
                envs.step(action.numpy())

            done = terminated | truncated

            obs_buf[step]     = obs
            act_buf[step]     = action
            logp_buf[step]    = log_prob
            rew_buf[step]     = torch.tensor(reward, dtype=torch.float32)
            done_buf[step]    = torch.tensor(done,   dtype=torch.float32)
            val_buf[step]     = value
            entropy_buf[step] = entropy

            ep_returns += reward
            for i, d in enumerate(done):
                if d:
                    episode_rewards.append(ep_returns[i])
                    ep_returns[i] = 0.0

            obs         = torch.tensor(next_obs, dtype=torch.float32)
            total_steps += num_envs

        # ── bootstrap ──────────────────────────────────────────
        with torch.no_grad():
            next_value = critic(obs).unsqueeze(0)   # (1, num_envs)

        advantages, returns = compute_gae(
            rew_buf, val_buf, done_buf,
            next_value, gamma, gae_lambda)

        # normalise advantages
        advantages = ((advantages - advantages.mean())
                      / (advantages.std() + 1e-8))

        # ── update ─────────────────────────────────────────────
        b_obs   = obs_buf.reshape(-1, obs_dim)
        b_act   = act_buf.reshape(-1, act_dim)
        b_adv   = advantages.reshape(-1)
        b_ret   = returns.reshape(-1)
        b_ent   = entropy_buf.reshape(-1)

        dist        = actor.forward(b_obs)
        new_logp    = dist.log_prob(b_act).sum(-1)
        new_entropy = dist.entropy().sum(-1)
        new_value   = critic(b_obs)

        actor_loss  = -(new_logp * b_adv.detach()).mean()
        critic_loss = 0.5 * (new_value - b_ret.detach()).pow(2).mean()
        entropy_loss= -new_entropy.mean()

        loss = actor_loss + vf_coef * critic_loss + ent_coef * entropy_loss

        optimizer.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(params, max_grad_norm)
        optimizer.step()

    envs.close()
    return episode_rewards


# ══════════════════════════════════════════════════════════════
# Plot 1 — GAE Advantage vs Timesteps for different λ
# ══════════════════════════════════════════════════════════════

T        = 30
num_envs = 1

rewards    = torch.sin(torch.linspace(0, 3.14, T)).unsqueeze(1)
values     = torch.cos(torch.linspace(0, 3.14, T)).unsqueeze(1) * 0.5
dones      = torch.zeros((T, 1))
dones[-1]  = 1.0
next_value = torch.zeros((1, 1))
gamma      = 0.99

lambdas        = np.linspace(0, 1, 6)
advantages_dict = {}

for lam in lambdas:
    adv, _ = compute_gae(
        rewards, values, dones, next_value, gamma, lam)
    advantages_dict[lam] = adv.squeeze().numpy()

plt.figure(figsize=(9, 5))
for lam in lambdas:
    plt.plot(advantages_dict[lam], label=f"λ={lam:.2f}")
plt.title("GAE advantage vs timestep for different λ")
plt.xlabel("Timestep")
plt.ylabel("Advantage")
plt.legend()
plt.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("gae_advantage_timesteps.pdf")
plt.show()
plt.close()


# ══════════════════════════════════════════════════════════════
# Plot 2 — GAE Bias-Variance Tradeoff
# ══════════════════════════════════════════════════════════════

var_list      = []
mean_abs_list = []
lambda_sweep  = np.linspace(0, 1, 20)

for lam in lambda_sweep:
    adv, _ = compute_gae(
        rewards, values, dones, next_value, gamma, lam)
    adv_np = adv.numpy()
    var_list.append(np.var(adv_np))
    mean_abs_list.append(np.mean(np.abs(adv_np)))

plt.figure(figsize=(9, 5))
plt.plot(lambda_sweep, var_list,      label="Variance")
plt.plot(lambda_sweep, mean_abs_list, label="Mean |Advantage|")
plt.title("GAE bias-variance tradeoff")
plt.xlabel("λ")
plt.ylabel("Metric")
plt.legend()
plt.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("gae_lambda_analysis.pdf")
plt.show()
plt.close()


# ══════════════════════════════════════════════════════════════
# Plot 3 — GAE λ Sweep on Hopper-v4
# ══════════════════════════════════════════════════════════════

env_id      = "Hopper-v4"
gae_lambdas = [0.0, 0.5, 0.8, 0.95, 1.0]
results     = {}

for lam in gae_lambdas:
    print(f"\n===== Training on {env_id} with λ={lam} =====")
    rewards = train_a2c(
        env_id          = env_id,
        gae_lambda      = lam,
        total_timesteps = 100_000,
        num_envs        = 4,
        num_steps       = 5,
        lr              = 3e-4,
    )
    results[lam] = rewards

plt.figure(figsize=(9, 5))
for lam, rews in results.items():
    rews = np.array(rews)
    if len(rews) == 0:
        continue
    plt.plot(moving_average(rews, 20), label=f"λ={lam}")
plt.title("GAE λ sweep on Hopper-v4")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.legend()
plt.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("gae_hopper_sweep.pdf")
plt.show()
plt.close()


# ══════════════════════════════════════════════════════════════
# Plot 4 — A2C vs A3C on Hopper-v4
# ══════════════════════════════════════════════════════════════

env_id = "Hopper-v4"

print("\n===== Training A2C on Hopper =====")
a2c_rewards = train_a2c(
    env_id          = env_id,
    total_timesteps = 100_000,
    num_envs        = 4,
    num_steps       = 5,
    lr              = 3e-4,
)

print("\n===== Training A3C on Hopper =====")
# FIX 5: train_a3c returns 5-tuple — unpack correctly
a3c_rewards, _, _, _, _ = train_a3c(
    env_id              = env_id,
    num_workers         = 4,
    episodes_per_worker = 50,
    gamma               = 0.99,
    num_steps           = 5,
    lr                  = 3e-5,
)

a2c_rewards = np.array(a2c_rewards)
a3c_rewards = np.array(a3c_rewards)

plt.figure(figsize=(9, 5))
plt.plot(moving_average(a2c_rewards, 20), label="A2C")
plt.plot(moving_average(a3c_rewards, 20), label="A3C")
plt.title("A2C vs A3C on Hopper-v4")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.legend()
plt.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("a2c_vs_a3c_hopper.pdf")
plt.show()
plt.close()
````

**Answer 1 :** In GAE, we start with multi-step advantage estimates, where we compute advantages using different step returns (1-step, 2-step, 3-step, etc.). Each of these gives a different estimate of how good an action is. GAE combines all these estimates into a single value by taking a weighted sum, where the weights decrease using a factor $\lambda$. This results in the GAE formula, which is essentially a sum of temporal difference (TD) errors multiplied by powers of $\lambda$. This way, it smoothly balances between short-term and long-term estimates.

**Answer 2 :** When $\lambda$ = 0, GAE reduces to using only the 1-step TD error, which has low variance but high bias. When $\lambda$ = 1, it becomes similar to using full returns, which has low bias but very high variance. In practice, a value between 0 and 1 is used to balance bias and variance. A commonly used default is $\lambda$ = 0.95, as it provides a good trade-off and stable learning.

**Answer 3 :** In the lambda sweep experiment, values around 0.9 to 0.97 usually perform the best. This is because they balance bias and variance effectively. Lower values (close to 0) rely too much on short-term estimates and may miss long-term rewards, while higher values (close to 1) introduce high variance and unstable learning. Hence, intermediate values give better and more stable performance.

**Answer 4 :** GAE is closely related to TD($\lambda$) from tabular reinforcement learning. Both methods use a weighted combination of multi-step returns with a decay factor $\lambda$. The main difference is that TD($\lambda$) is used to estimate value functions, while GAE is used to estimate advantages in policy gradient methods. So, GAE can be seen as an extension of TD($\lambda$) applied in the actor-critic setting.

````python
class DDPG():
    def __init__(self, env_name, seed, gamma, tau,
                 bufferSize, batch_size, updateFrequency,
                 policyOptimizerFn, valueOptimizerFn,
                 policyOptimizerLR, valueOptimizerLR,
                 MAX_TRAIN_EPISODES, MAX_EVAL_EPISODE):
        #this DDPG method
        # 1. creates and initializes (with seed) the environment, train/eval episodes, gamma, etc.
        # 2. creates and intializes all the variables required for book-keeping values via the initBookKeeping method
        # 3. creates targetValueNetwork , targetPolicyNetwork
        # 4. creates and initializes (with network params) the optimizer function
        # 5. creates onlineValueNetwork, onlinePolicyNetwork
        # 6. Creates the replayBuffer

        #Your code goes in here
        

        self.env = gym.make(env_name)
        self.env.reset(seed=seed)
        torch.manual_seed(seed)
        np.random.seed(seed)
        random.seed(seed)

        self.gamma           = gamma
        self.tau             = tau
        self.bufferSize      = bufferSize
        self.batch_size      = batch_size
        self.updateFrequency = updateFrequency
        self.MAX_TRAIN_EPISODES = MAX_TRAIN_EPISODES
        self.MAX_EVAL_EPISODE   = MAX_EVAL_EPISODE

        obs_dim      = self.env.observation_space.shape[0]
        act_dim      = self.env.action_space.shape[0]
        action_scale = float(self.env.action_space.high[0])

        
        self.onlinePolicyNetwork = DDPGPolicyNetwork(obs_dim, act_dim, action_scale)
        self.targetPolicyNetwork = DDPGPolicyNetwork(obs_dim, act_dim, action_scale)
        self.onlineValueNetwork  = DDPGValueNetwork(obs_dim, act_dim)
        self.targetValueNetwork  = DDPGValueNetwork(obs_dim, act_dim)

        self.targetPolicyNetwork.load_state_dict(
            self.onlinePolicyNetwork.state_dict())
        self.targetValueNetwork.load_state_dict(
            self.onlineValueNetwork.state_dict())

        for p in self.targetPolicyNetwork.parameters():
            p.requires_grad = False
        for p in self.targetValueNetwork.parameters():
            p.requires_grad = False

        self.policyOptimizer = policyOptimizerFn(
            self.onlinePolicyNetwork.parameters(), lr=policyOptimizerLR)
        self.valueOptimizer  = valueOptimizerFn(
            self.onlineValueNetwork.parameters(),  lr=valueOptimizerLR)

        self.replayBuffer = deque(maxlen=bufferSize)
        self.startTime    = time.time()
        self.initBookKeeping()

    def initBookKeeping(self):
        self.trainRewards = []
        self.evalRewards  = []
        self.steps        = []
        self.trainTimes   = []
        self.wallTimes    = []


````

````python
class DDPG(DDPG):
    def updateNetworks(self, onlineNet, targetNet, tau):
        #this function updates the onlineNetwork with the target network
        #
        # Your code goes in here
        #
        for t_param, o_param in zip(targetNet.parameters(),
                                    onlineNet.parameters()):
            t_param.data.copy_(
                tau * o_param.data + (1 - tau) * t_param.data)
````

````python
class DDPG(DDPG):
    def gaussianStrategy(self, net, s, envActionRange,
                         noiseScaleRatio, explorationMax=False):
        #this function sets the scale of exploration then add the noise of this scale to the greedy action
        #and clips it within the range

        #Your code here
        
        actionLowVal, actionHighVal = envActionRange
        s_t          = torch.tensor(s, dtype=torch.float32).unsqueeze(0)
        greedyAction = net(s_t).detach().numpy()[0]
        scale        = float(actionHighVal[0])
        if not explorationMax:
            scale = noiseScaleRatio * scale
        noise  = np.random.normal(0, scale, size=greedyAction.shape)
        action = greedyAction + noise
        return np.clip(action, actionLowVal, actionHighVal)
````

````python
class DDPG(DDPG):
    def greedyStrategy(self, net, s, envActionRange):
        #this function selects the greedy action
        #and clips it within the range

        #Your code here
        
        actionLowVal, actionHighVal = envActionRange
        s_t    = torch.tensor(s, dtype=torch.float32).unsqueeze(0)
        action = net(s_t).detach().numpy()[0]
        return np.clip(action, actionLowVal, actionHighVal)
````

````python
class DDPG(DDPG):
    def trainAgent(self):
        #this method collects experiences and trains the agent and does BookKeeping while training.
        #this calls the trainNetwork() method internally, it also evaluates the agent per episode
        #it trains the agent for MAX_TRAIN_EPISODES
        #
        #Your code goes in here
        #
        
        trainRewardsList  = []
        evalRewardsList   = []
        trainTimeList     = []
        wallClockTimeList = []
        stepsList         = []

        train_elapsed  = 0.0
        envActionRange = (self.env.action_space.low,
                          self.env.action_space.high)

        for ep in range(self.MAX_TRAIN_EPISODES):
            s, _      = self.env.reset()
            done      = False
            truncated = False
            ep_reward = 0
            ep_steps  = 0
            ep_start  = time.time()

            while not (done or truncated):
                exploration_max = len(self.replayBuffer) < self.batch_size
                a = self.gaussianStrategy(
                    self.onlinePolicyNetwork, s, envActionRange,
                    noiseScaleRatio=0.1,
                    explorationMax=exploration_max)

                s_next, r, done, truncated, _ = self.env.step(a)
                self.replayBuffer.append((s, a, r, s_next,
                                          float(done and not truncated)))
                s          = s_next
                ep_reward += r
                ep_steps  += 1

                if len(self.replayBuffer) >= self.batch_size:
                    batch = random.sample(self.replayBuffer, self.batch_size)
                    self.trainNetwork(batch)

            if (ep + 1) % self.updateFrequency == 0:
                self.updateNetworks(
                    self.onlinePolicyNetwork, self.targetPolicyNetwork, self.tau)
                self.updateNetworks(
                    self.onlineValueNetwork,  self.targetValueNetwork,  self.tau)

            train_elapsed += time.time() - ep_start

            trainRewardsList.append(ep_reward)
            stepsList.append(ep_steps)
            trainTimeList.append(train_elapsed)
            wallClockTimeList.append(time.time() - self.startTime)

            if (ep + 1) % 10 == 0:
                eval_r = self.evaluateAgent()
                evalRewardsList.append((ep + 1, eval_r))
                print(f"  Ep {ep+1:4d} | train {ep_reward:8.1f}"
                      f" | eval {eval_r:8.1f}")

        return (trainRewardsList, trainTimeList,
                evalRewardsList, wallClockTimeList, stepsList)


        
````

````python
class DDPG(DDPG):
    def trainNetwork(self, experiences):
        # this method trains the value network epoch number of times and is called by the trainAgent function
        # it essentially uses the experiences to calculate target, using the targets it calculates the error, which
        # is further used for calulating the loss. It then uses the optimizer over the loss
        # to update the params of the network by backpropagating through the network
        # this function does not return anything
        # you can try out other loss functions other than MSE like Huber loss, MAE, etc.
        #
        #Your code goes in here
        #

        
        states      = torch.tensor(
            np.array([e[0] for e in experiences]), dtype=torch.float32)
        actions     = torch.tensor(
            np.array([e[1] for e in experiences]), dtype=torch.float32)
        rewards     = torch.tensor(
            np.array([e[2] for e in experiences]),
            dtype=torch.float32).unsqueeze(1)
        next_states = torch.tensor(
            np.array([e[3] for e in experiences]), dtype=torch.float32)
        dones       = torch.tensor(
            np.array([e[4] for e in experiences]),
            dtype=torch.float32).unsqueeze(1)

        with torch.no_grad():
            next_actions = self.targetPolicyNetwork(next_states)
            target_q     = self.targetValueNetwork(next_states, next_actions)
            target       = rewards + self.gamma * (1 - dones) * target_q

        current_q  = self.onlineValueNetwork(states, actions)
        valueLoss  = F.mse_loss(current_q, target)

        self.valueOptimizer.zero_grad()
        valueLoss.backward()
        torch.nn.utils.clip_grad_norm_(
            self.onlineValueNetwork.parameters(), 1.0)
        self.valueOptimizer.step()

        pred_actions = self.onlinePolicyNetwork(states)
        policyLoss   = -self.onlineValueNetwork(states, pred_actions).mean()

        self.policyOptimizer.zero_grad()
        policyLoss.backward()
        torch.nn.utils.clip_grad_norm_(
            self.onlinePolicyNetwork.parameters(), 1.0)
        self.policyOptimizer.step()
````

````python
class DDPG(DDPG):
    def evaluateAgent(self):
        #this function evaluates the agent using the value network, it evaluates agent for MAX_EVAL_EPISODES
        #typcially MAX_EVAL_EPISODES = 1
        #
        #Your code goes in here
        #
        
        rewards        = []
        envActionRange = (self.env.action_space.low,
                          self.env.action_space.high)

        for _ in range(self.MAX_EVAL_EPISODE):
            s, _      = self.env.reset()
            done      = False
            truncated = False
            total     = 0

            while not (done or truncated):
                a = self.greedyStrategy(
                    self.onlinePolicyNetwork, s, envActionRange)
                s, r, done, truncated, _ = self.env.step(a)
                total += r

            rewards.append(total)

        return float(np.mean(rewards))

        
````

# Soft Actor-Critic (SAC) [ 50 marks ]

SAC is an off-policy maximum entropy deep RL algorithm for continuous action spaces. It maintains two online Q-networks and two target Q-networks to mitigate overestimation bias, and uses a squashed Gaussian policy with automatic entropy tuning.

### Key Ideas

**1. Stochastic Policy (Squashed Gaussian)**  
The network learns $\mu$ and $\sigma$; samples $u \sim \mathcal{N}(\mu, \sigma)$, applies $a = \tanh(u)$, then rescales to the environment action range.

**2. Log-probability correction for tanh squashing:**

$$\log \pi(a \mid s) = \log p(u \mid s) - \sum_{i=0}^{D} \log(1 - \tanh^2(u_i))$$

**3. Policy Loss (maximize Q, maximize entropy):**

$$J_\pi(\phi) = -\, \mathbb{E}_{s \sim \mathcal{U}(D),\, \hat{a} \sim \pi} \left[ \min_k Q(s, \hat{a};\, \theta^k) - \alpha \log \pi(\hat{a} \mid s;\, \phi) \right]$$

**4. Automatic Entropy Tuning (learn $\alpha$):**

$$J(\alpha) = -\, \mathbb{E}_{s \sim \mathcal{U}(D),\, \hat{a} \sim \pi} \left[ \alpha \left( \mathcal{H} + \log \pi(\hat{a} \mid s;\, \phi) \right) \right]$$

where $\mathcal{H} = -|\mathcal{A}| = -\prod \text{shape}(a_\text{env})$

**5. Soft Bellman Backup (critic target):**

$$\text{target\_q} = r + \gamma \cdot \left( \min_k Q_{\text{target},k}(s', a') - \alpha \log \pi(a' \mid s') \right) \cdot (1 - \text{done})$$

**6. Soft (Polyak) Target Update:**

$$\theta_{\text{target}} \leftarrow (1 - \tau)\, \theta_{\text{target}} + \tau\, \theta_{\text{online}}$$



### Task

Implement the SAC agent using the class structure below.

````python
import torch
import torch.nn as nn
import numpy as np
import gymnasium as gym
import random
import time
from collections import deque

# ── ValueNetwork ──────────────────────────────────────────────
class SACValueNetwork(nn.Module):

    def __init__(self, stateDim, actionDim, hiddenDims):
        super().__init__()
        layers    = []
        input_dim = stateDim + actionDim
        for h in hiddenDims:
            layers.append(nn.Linear(input_dim, h))
            layers.append(nn.ReLU())
            input_dim = h
        layers.append(nn.Linear(input_dim, 1))
        self.net = nn.Sequential(*layers)

    def forward(self, state, action):
        x = torch.cat([state, action], dim=-1)
        return self.net(x)


# ── PolicyNetwork ─────────────────────────────────────────────
class SACPolicyNetwork(nn.Module):

    def __init__(self, stateDim, actionRange, hiddenDims):
        super().__init__()
        self.low  = torch.tensor(actionRange[0], dtype=torch.float32)
        self.high = torch.tensor(actionRange[1], dtype=torch.float32)

        layers    = []
        input_dim = stateDim
        for h in hiddenDims:
            layers.append(nn.Linear(input_dim, h))
            layers.append(nn.ReLU())
            input_dim = h

        self.net         = nn.Sequential(*layers)
        self.mean_layer  = nn.Linear(input_dim, len(self.high))
        self.log_std_layer = nn.Linear(input_dim, len(self.high))

        self.log_std_min = -20
        self.log_std_max = 2

        # FIX 1: log_alpha as nn.Parameter so it moves with the model
        self.log_alpha = nn.Parameter(torch.zeros(1))

    def forward(self, state):
        x       = self.net(state)
        mean    = self.mean_layer(x)
        log_std = torch.clamp(
            self.log_std_layer(x), self.log_std_min, self.log_std_max)
        std     = log_std.exp()

        dist    = torch.distributions.Normal(mean, std)
        u       = dist.rsample()
        a       = torch.tanh(u)

        action_env    = self.rescale(a)
        action_greedy = self.rescale(torch.tanh(mean))

        # FIX 2: squeeze log_prob to (batch,) not (batch,1)
        # keepdim=False avoids shape mismatch in alpha loss
        log_prob = (dist.log_prob(u)
                    - torch.log(1 - a.pow(2) + 1e-6)
                   ).sum(dim=-1, keepdim=True)

        return action_env, action_greedy, log_prob

    def rescale(self, action):
        return self.low + (action + 1.0) * 0.5 * (self.high - self.low)
````

````python
class ReplayBuffer:

    def __init__(self, bufferSize):
        # FIX 3: use deque for O(1) append/pop instead of list.pop(0)
        self.buffer   = deque(maxlen=bufferSize)
        self.max_size = bufferSize

    def store(self, transition):
        self.buffer.append(transition)

    def sample(self, batchSize):
        return random.sample(self.buffer, batchSize)
````

````python



class SAC:

    def __init__(self, env, gamma, tau, bufferSize,
                 entropyLR,
                 updateFrequency,
                 policyOptimizerFn,
                 valueOptimizerFn_1,
                 valueOptimizerFn_2,
                 policyOptimizerLR,
                 valueOptimizerLR,
                 alphaOptimizerFn,
                 MAX_TRAIN_EPISODES,
                 MAX_EVAL_EPISODE,
                 hDims,
                 batchSize,
                 minSamples,
                 seed=0):

        self.env               = env
        self.gamma             = gamma
        self.tau               = tau
        self.batchSize         = batchSize
        self.updateFrequency   = updateFrequency
        self.MAX_TRAIN_EPISODES= MAX_TRAIN_EPISODES
        self.MAX_EVAL_EPISODE  = MAX_EVAL_EPISODE
        self.minSamples        = minSamples

        # seeding
        torch.manual_seed(seed)
        np.random.seed(seed)
        random.seed(seed)
        self.env.reset(seed=seed)

        stateDim    = env.observation_space.shape[0]
        actionDim   = env.action_space.shape[0]
        actionRange = (env.action_space.low, env.action_space.high)

        # FIX 4: use renamed network classes to avoid collision
        self.q1        = SACValueNetwork(stateDim, actionDim, hDims)
        self.q2        = SACValueNetwork(stateDim, actionDim, hDims)
        self.q1_target = SACValueNetwork(stateDim, actionDim, hDims)
        self.q2_target = SACValueNetwork(stateDim, actionDim, hDims)
        self.policy    = SACPolicyNetwork(stateDim, actionRange, hDims)

        self.q1_target.load_state_dict(self.q1.state_dict())
        self.q2_target.load_state_dict(self.q2.state_dict())

        # freeze target networks
        for p in self.q1_target.parameters():
            p.requires_grad = False
        for p in self.q2_target.parameters():
            p.requires_grad = False

        self.q1_opt     = valueOptimizerFn_1(
            self.q1.parameters(),     lr=valueOptimizerLR)
        self.q2_opt     = valueOptimizerFn_2(
            self.q2.parameters(),     lr=valueOptimizerLR)
        self.policy_opt = policyOptimizerFn(
            self.policy.parameters(), lr=policyOptimizerLR)
        self.alpha_opt  = alphaOptimizerFn(
            [self.policy.log_alpha],  lr=entropyLR)

        self.buffer    = ReplayBuffer(bufferSize)
        self.startTime = time.time()

    def selectAction(self, state):
        state          = torch.tensor(
            state, dtype=torch.float32).unsqueeze(0)
        action_env, _, _ = self.policy(state)
        return action_env.detach().numpy()[0]

    def trainNetworks(self, batch):
        s, a, r, sn, d = zip(*batch)

        s  = torch.tensor(np.array(s),  dtype=torch.float32)
        a  = torch.tensor(np.array(a),  dtype=torch.float32)
        r  = torch.tensor(np.array(r),  dtype=torch.float32).unsqueeze(1)
        sn = torch.tensor(np.array(sn), dtype=torch.float32)
        d  = torch.tensor(np.array(d),  dtype=torch.float32).unsqueeze(1)

        alpha = self.policy.log_alpha.exp().detach()

        with torch.no_grad():
            next_a, _, logp_next = self.policy(sn)
            q1_t     = self.q1_target(sn, next_a)
            q2_t     = self.q2_target(sn, next_a)
            q_target = r + self.gamma * (
                torch.min(q1_t, q2_t) - alpha * logp_next) * (1 - d)

        # ── critic update ─────────────────────────────────────
        q1_val   = self.q1(s, a)
        q2_val   = self.q2(s, a)
        loss_q1  = ((q1_val - q_target) ** 2).mean()
        loss_q2  = ((q2_val - q_target) ** 2).mean()

        self.q1_opt.zero_grad()
        loss_q1.backward()
        self.q1_opt.step()

        self.q2_opt.zero_grad()
        loss_q2.backward()
        self.q2_opt.step()

        # ── policy update ─────────────────────────────────────
        a_new, _, logp_new = self.policy(s)
        q1_pi  = self.q1(s, a_new)
        q2_pi  = self.q2(s, a_new)

        policy_loss = (alpha * logp_new
                       - torch.min(q1_pi, q2_pi)).mean()

        self.policy_opt.zero_grad()
        policy_loss.backward()
        self.policy_opt.step()

        # ── alpha (entropy) update ────────────────────────────
        target_entropy = -float(a.shape[1])
        # FIX 5: detach logp_new so grad only flows through log_alpha
        alpha_loss = -(self.policy.log_alpha
                       * (logp_new.detach() + target_entropy)).mean()

        self.alpha_opt.zero_grad()
        alpha_loss.backward()
        self.alpha_opt.step()

    def updateTargetNetworks(self):
        for t, o in zip(self.q1_target.parameters(),
                        self.q1.parameters()):
            t.data.copy_((1 - self.tau) * t.data + self.tau * o.data)
        for t, o in zip(self.q2_target.parameters(),
                        self.q2.parameters()):
            t.data.copy_((1 - self.tau) * t.data + self.tau * o.data)

    def trainAgent(self):
        trainRewards  = []
        evalRewards   = []
        stepsList     = []
        trainTimeList = []
        wallTimeList  = []

        train_elapsed  = 0.0

        # FIX 6: use MAX_TRAIN_EPISODES not hardcoded 100
        for ep in range(self.MAX_TRAIN_EPISODES):
            s, _      = self.env.reset()
            done      = False
            truncated = False    # FIX 7: track truncated
            total     = 0
            ep_steps  = 0
            ep_start  = time.time()

            while not (done or truncated):
                # FIX 8: use minSamples for warm-up threshold
                if len(self.buffer.buffer) < self.minSamples:
                    a = self.env.action_space.sample()
                else:
                    a = self.selectAction(s)

                sn, r, done, truncated, _ = self.env.step(a)
                # FIX 9: only mark terminal if truly done not truncated
                self.buffer.store((s, a, r, sn,
                                   float(done and not truncated)))
                s        = sn
                total   += r
                ep_steps += 1

                if len(self.buffer.buffer) >= self.minSamples:
                    batch = self.buffer.sample(self.batchSize)
                    self.trainNetworks(batch)

                    # FIX 10: soft update every updateFrequency steps
                    if ep_steps % self.updateFrequency == 0:
                        self.updateTargetNetworks()

            train_elapsed += time.time() - ep_start
            trainRewards.append(total)
            stepsList.append(ep_steps)
            trainTimeList.append(train_elapsed)
            wallTimeList.append(time.time() - self.startTime)

            if (ep + 1) % 10 == 0:
                eval_r = self.evaluateAgent()
                evalRewards.append((ep + 1, eval_r))
                print(f"  Ep {ep+1:4d} | train {total:8.1f}"
                      f" | eval {eval_r:8.1f}")

        return trainRewards, evalRewards, stepsList, trainTimeList, wallTimeList

    def evaluateAgent(self):
        rewards = []
        for _ in range(self.MAX_EVAL_EPISODE):
            s, _      = self.env.reset()
            done      = False
            truncated = False
            total     = 0
            while not (done or truncated):
                # FIX 11: use greedy action (index 1) correctly
                with torch.no_grad():
                    s_t = torch.tensor(
                        s, dtype=torch.float32).unsqueeze(0)
                    _, a_greedy, _ = self.policy(s_t)
                s, r, done, truncated, _ = self.env.step(
                    a_greedy.detach().numpy()[0])
                total += r
            rewards.append(total)
        return float(np.mean(rewards))

    def runSAC(self):
        # FIX 12: return 4-tuple to match calling code unpacking
        (trainRewards, evalRewards,
         stepsList, trainTimeList,
         wallTimeList) = self.trainAgent()

        return (trainRewards, evalRewards,
                trainTimeList, wallTimeList)


````

````python
import os
import random
import time
import torch.multiprocessing as mp

import gymnasium as gym
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.distributions import Normal
import matplotlib.pyplot as plt

# ── Hyperparameters ───────────────────────────────────────────────────────────
ppo_gym_id          = "HalfCheetah-v4"
ppo_learning_rate   = 3e-4
ppo_seed            = 1
ppo_total_timesteps = 1_000_000

ppo_num_workers     = 4
ppo_num_steps       = 128

ppo_gamma           = 0.99
ppo_gae_lambda      = 0.95

ppo_update_epochs   = 10
ppo_num_minibatches = 32

ppo_clip_coef       = 0.2
ppo_clip_vloss      = True

ppo_ent_coef        = 0.0
ppo_vf_coef         = 0.5
ppo_max_grad_norm   = 0.5

ppo_norm_adv        = True
ppo_anneal_lr       = True
ppo_target_kl       = 0.01

ppo_batch_size     = ppo_num_workers * ppo_num_steps
ppo_minibatch_size = ppo_batch_size // ppo_num_minibatches

print(f"PPO: batch={ppo_batch_size}, minibatch={ppo_minibatch_size}")

````

````python
# PPO Agent (takes raw dims, works in any process)
class PPOAgent(nn.Module):
    """
    PPO agent containing separate actor and critic networks.
    Supports both discrete and continuous action spaces.
    """

    def __init__(self, obs_dim, act_dim, is_discrete=False,
                 action_high=None, action_low=None):
        """
        Initialize the PPO agent.

        Args:
            obs_dim (int): Dimension of the observation/state space.
            act_dim (int): Dimension of the action space.
            is_discrete (bool): True if the environment has a discrete
                                action space, otherwise False.
            action_high (array-like, optional): Upper bound of the
                                continuous action space.
            action_low (array-like, optional): Lower bound of the
                                continuous action space.

        Expected Output:
            Initializes actor and critic neural networks along with
            parameters required for action scaling in continuous spaces.
        """

        #Your Code goes here
        

    
        super().__init__()

        # FIX 1: separate actor/critic with larger networks for HalfCheetah
        self.actor_mean = nn.Sequential(
            nn.Linear(obs_dim, 256), nn.Tanh(),
            nn.Linear(256, 256),     nn.Tanh(),
            nn.Linear(256, act_dim)
        )
        # FIX 2: log_std as learned parameter, not fixed ones
        self.actor_log_std = nn.Parameter(
            torch.zeros(act_dim)
        )

        self.critic = nn.Sequential(
            nn.Linear(obs_dim, 256), nn.Tanh(),
            nn.Linear(256, 256),     nn.Tanh(),
            nn.Linear(256, 1)
        )

        # orthogonal init (standard PPO practice)
        for layer in self.actor_mean:
            if isinstance(layer, nn.Linear):
                nn.init.orthogonal_(layer.weight, gain=np.sqrt(2))
                nn.init.zeros_(layer.bias)
        nn.init.orthogonal_(self.actor_mean[-1].weight, gain=0.01)

        for layer in self.critic:
            if isinstance(layer, nn.Linear):
                nn.init.orthogonal_(layer.weight, gain=np.sqrt(2))
                nn.init.zeros_(layer.bias)
        nn.init.orthogonal_(self.critic[-1].weight, gain=1.0)

    def get_value(self, x):
        return self.critic(x)

    def get_action_and_value(self, x, action=None):
        mean    = self.actor_mean(x)
        # FIX 3: std is exp of learned log_std, broadcast over batch
        std     = self.actor_log_std.exp().expand_as(mean)
        dist    = Normal(mean, std)

        if action is None:
            action = dist.sample()

        # FIX 4: sum log_prob over action dims (per-dim → scalar per step)
        log_prob = dist.log_prob(action).sum(dim=-1)
        entropy  = dist.entropy().sum(dim=-1)
        value    = self.critic(x)

        return action, log_prob, entropy, value


        

        
````

````python
# PPO Worker + Training Loop (Multiprocessing Version)
def ppo_worker(worker_id, global_agent,
               env_id, seed, num_steps, num_updates,
               shared_obs, shared_actions, shared_logprobs,
               shared_rewards, shared_dones, shared_values,
               shared_next_obs, shared_next_done,
               collect_barrier, update_barrier, ep_queue):
    """
    PPO worker process responsible for collecting rollouts.

    Args:
        worker_id        : Unique ID of the worker process.
        global_agent     : Shared PPO policy network.
        env_id           : Gym environment name.
        seed             : Random seed for reproducibility.
        num_steps        : Number of steps collected per rollout.
        num_updates      : Number of PPO updates to perform.

        shared_obs       : Shared buffer storing observations.
        shared_actions   : Shared buffer storing actions.
        shared_logprobs  : Shared buffer storing log probabilities.
        shared_rewards   : Shared buffer storing rewards.
        shared_dones     : Shared buffer storing termination flags.
        shared_values    : Shared buffer storing value predictions.

        shared_next_obs  : Shared buffer storing next observations.
        shared_next_done : Shared buffer storing next done flag.

        collect_barrier  : Synchronization barrier for rollout collection.
        update_barrier   : Synchronization barrier after PPO update.
        ep_queue         : Queue used to send episode rewards to main process.

    Expected Output:
        This function does not return values directly.
        Instead, it stores rollout data in shared buffers and
        sends episode rewards to the main process through ep_queue.
    """

    

    env     = gym.make(env_id)
    obs, _  = env.reset(seed=seed + worker_id)

    # FIX 5: track episode return properly
    ep_return = 0.0

    for update in range(num_updates):
        for step in range(num_steps):
            obs_tensor = torch.tensor(obs, dtype=torch.float32)

            with torch.no_grad():
                action, logprob, _, value = \
                    global_agent.get_action_and_value(obs_tensor)

            # FIX 6: continuous action — use numpy directly, not .item()
            act = action.numpy()

            next_obs, reward, done, truncated, _ = env.step(act)
            ep_return += reward

            shared_obs[worker_id, step]      = obs_tensor
            shared_actions[worker_id, step]  = action
            shared_logprobs[worker_id, step] = logprob
            shared_rewards[worker_id, step]  = reward
            # FIX 7: mark done for either terminal or truncated
            shared_dones[worker_id, step]    = float(done or truncated)
            shared_values[worker_id, step]   = value.squeeze()

            obs = next_obs

            if done or truncated:
                # FIX 8: put full episode return, not just last reward
                ep_queue.put(ep_return)
                ep_return = 0.0
                obs, _ = env.reset()

        shared_next_obs[worker_id]  = torch.tensor(obs, dtype=torch.float32)
        # FIX 9: next_done reflects whether current obs is from reset
        shared_next_done[worker_id] = float(done or truncated)

        collect_barrier.wait()
        update_barrier.wait()

````

````python
def train_ppo(env_id=ppo_gym_id, seed=ppo_seed,
              total_timesteps=ppo_total_timesteps,
              num_workers=ppo_num_workers, num_steps=ppo_num_steps,
              lr=ppo_learning_rate, gamma=ppo_gamma,
              gae_lambda=ppo_gae_lambda, update_epochs=ppo_update_epochs,
              num_minibatches=ppo_num_minibatches,
              clip_coef=ppo_clip_coef, clip_vloss=ppo_clip_vloss,
              ent_coef=ppo_ent_coef, vf_coef=ppo_vf_coef,
              max_grad_norm=ppo_max_grad_norm, norm_adv=ppo_norm_adv,
              anneal_lr=ppo_anneal_lr, target_kl=ppo_target_kl):
    """
    PPO training loop using multiprocessing workers.

    Args:
        env_id           : Gym environment used for training.
        seed             : Random seed.
        total_timesteps  : Total number of environment interactions.
        num_workers      : Number of parallel workers collecting data.
        num_steps        : Number of rollout steps per worker.

        lr               : Learning rate for optimizer.
        gamma            : Discount factor for future rewards.
        gae_lambda       : Lambda parameter for GAE.

        update_epochs    : Number of PPO update passes per batch.
        num_minibatches  : Number of minibatches for SGD.

        clip_coef        : PPO clipping coefficient.
        clip_vloss       : Whether value loss clipping is used.

        ent_coef         : Entropy coefficient for exploration.
        vf_coef          : Value loss coefficient.
        max_grad_norm    : Maximum gradient norm for clipping.

        norm_adv         : Whether to normalize advantages.
        anneal_lr        : Whether learning rate is annealed.
        target_kl        : Optional KL divergence stopping threshold.

    Expected Output:
        results      : List containing tuples (global_step, episode_reward)
                       used to plot the learning curve.
        global_agent : The trained PPO agent.
    """



    

    # seeding
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    env     = gym.make(env_id)
    obs_dim = env.observation_space.shape[0]

    # FIX 10: HalfCheetah is continuous — use shape[0] not .n
    act_dim = env.action_space.shape[0]
    env.close()

    batch_size     = num_workers * num_steps
    minibatch_size = batch_size // num_minibatches
    num_updates    = total_timesteps // batch_size

    agent     = PPOAgent(obs_dim, act_dim)
    agent.share_memory()     # required for multiprocessing
    optimizer = optim.Adam(agent.parameters(), lr=lr, eps=1e-5)

    # shared tensors
    shared_obs       = torch.zeros((num_workers, num_steps, obs_dim))
    shared_actions   = torch.zeros((num_workers, num_steps, act_dim))
    shared_logprobs  = torch.zeros((num_workers, num_steps))
    shared_rewards   = torch.zeros((num_workers, num_steps))
    shared_dones     = torch.zeros((num_workers, num_steps))
    shared_values    = torch.zeros((num_workers, num_steps))
    shared_next_obs  = torch.zeros((num_workers, obs_dim))
    shared_next_done = torch.zeros((num_workers,))

    for t in [shared_obs, shared_actions, shared_logprobs,
              shared_rewards, shared_dones, shared_values,
              shared_next_obs, shared_next_done]:
        t.share_memory_()

    collect_barrier = mp.Barrier(num_workers + 1)
    update_barrier  = mp.Barrier(num_workers + 1)
    ep_queue        = mp.Queue()

    workers = []
    for i in range(num_workers):
        p = mp.Process(target=ppo_worker, args=(
            i, agent, env_id, seed, num_steps, num_updates,
            shared_obs, shared_actions, shared_logprobs,
            shared_rewards, shared_dones, shared_values,
            shared_next_obs, shared_next_done,
            collect_barrier, update_barrier, ep_queue
        ))
        p.daemon = True
        p.start()
        workers.append(p)

    # book-keeping
    train_rewards    = []
    eval_rewards_log = []
    train_times      = []
    wall_times       = []
    steps_log        = []

    start_time    = time.time()
    train_elapsed = 0.0
    global_step   = 0

    for update in range(1, num_updates + 1):

        # FIX 11: LR annealing
        if anneal_lr:
            frac = 1.0 - (update - 1.0) / num_updates
            optimizer.param_groups[0]["lr"] = frac * lr

        collect_barrier.wait()
        ep_start = time.time()

        with torch.no_grad():
            next_value = agent.get_value(shared_next_obs).squeeze()

        # FIX 12: GAE using lastgaelam carry variable (not indexing t+1)
        advantages   = torch.zeros_like(shared_rewards)
        lastgaelam   = 0.0

        for t in reversed(range(num_steps)):
            if t == num_steps - 1:
                next_nonterminal = 1.0 - shared_next_done
                next_vals        = next_value
            else:
                next_nonterminal = 1.0 - shared_dones[:, t + 1]
                next_vals        = shared_values[:, t + 1]

            delta      = (shared_rewards[:, t]
                          + gamma * next_vals * next_nonterminal
                          - shared_values[:, t])
            lastgaelam = delta + gamma * gae_lambda * next_nonterminal * lastgaelam
            advantages[:, t] = lastgaelam

        returns = advantages + shared_values

        # flatten batch
        b_obs       = shared_obs.reshape(-1, obs_dim)
        b_actions   = shared_actions.reshape(-1, act_dim)
        b_logprobs  = shared_logprobs.reshape(-1)
        b_advantages= advantages.reshape(-1)
        b_returns   = returns.reshape(-1)
        # FIX 13: collect b_values for clipped value loss
        b_values    = shared_values.reshape(-1)

        if norm_adv:
            b_advantages = ((b_advantages - b_advantages.mean())
                            / (b_advantages.std() + 1e-8))

        # PPO update epochs
        approx_kl_list = []

        for epoch in range(update_epochs):
            indices = np.random.permutation(batch_size)

            for start in range(0, batch_size, minibatch_size):
                mb_idx = indices[start:start + minibatch_size]

                _, newlogprob, entropy, new_value = \
                    agent.get_action_and_value(
                        b_obs[mb_idx], b_actions[mb_idx])

                log_ratio    = newlogprob - b_logprobs[mb_idx]
                ratio        = log_ratio.exp()

                # approximate KL for early stopping
                with torch.no_grad():
                    approx_kl = ((ratio - 1) - log_ratio).mean()
                    approx_kl_list.append(approx_kl.item())

                mb_adv = b_advantages[mb_idx]

                pg_loss1     = -mb_adv * ratio
                pg_loss2     = -mb_adv * torch.clamp(
                    ratio, 1 - clip_coef, 1 + clip_coef)
                policy_loss  = torch.max(pg_loss1, pg_loss2).mean()

                # FIX 14: clipped value loss (was completely missing)
                new_value = new_value.squeeze()
                if clip_vloss:
                    v_clipped  = (b_values[mb_idx]
                                  + torch.clamp(new_value - b_values[mb_idx],
                                                -clip_coef, clip_coef))
                    vf_loss1   = (new_value    - b_returns[mb_idx]).pow(2)
                    vf_loss2   = (v_clipped    - b_returns[mb_idx]).pow(2)
                    value_loss = 0.5 * torch.max(vf_loss1, vf_loss2).mean()
                else:
                    value_loss = 0.5 * (new_value - b_returns[mb_idx]).pow(2).mean()

                entropy_loss = entropy.mean()
                loss = (policy_loss
                        + vf_coef * value_loss
                        - ent_coef * entropy_loss)

                optimizer.zero_grad()
                loss.backward()
                nn.utils.clip_grad_norm_(agent.parameters(), max_grad_norm)
                optimizer.step()

            # FIX 15: early stopping on KL divergence
            if target_kl is not None:
                if np.mean(approx_kl_list) > target_kl:
                    break

        train_elapsed += time.time() - ep_start
        global_step   += batch_size

        # collect episode returns from queue
        ep_returns = []
        while not ep_queue.empty():
            ep_returns.append(ep_queue.get_nowait())

        mean_ep_return = np.mean(ep_returns) if ep_returns else np.nan

        # FIX 16: actually store results (was always empty list)
        train_rewards.append(mean_ep_return)
        steps_log.append(global_step)
        train_times.append(train_elapsed)
        wall_times.append(time.time() - start_time)

        if update % 10 == 0:
            print(f"  Update {update:4d}/{num_updates} | "
                  f"steps {global_step:8d} | "
                  f"mean_ep_ret {mean_ep_return:8.1f}")

        update_barrier.wait()

    for p in workers:
        p.join()

    return (train_rewards, steps_log,
            train_times, wall_times, agent)



````

````python
import numpy as np
import matplotlib.pyplot as plt
import gymnasium as gym
import torch
import torch.optim as optim

# ══════════════════════════════════════════════════════════════
# Configuration
# ══════════════════════════════════════════════════════════════

SEEDS         = [0, 1, 2, 3, 4]
EVAL_INTERVAL = 10

# Agent colors — distinct, colorblind-friendly
COLORS = {
    "REINFORCE" : "#e41a1c",   # red
    "VPG"       : "#4daf4a",   # green
    "DDPG"      : "#ff7f00",   # orange
    "TD3"       : "#984ea3",   # purple
    "PPO"       : "#008080",   # teal
}

# ── Helpers ───────────────────────────────────────────────────
def rolling_mean(arr, window=20):
    result = np.zeros_like(arr, dtype=float)
    for i in range(arr.shape[1]):
        lo = max(0, i - window + 1)
        result[:, i] = arr[:, lo:i+1].mean(axis=1)
    return result

def plot_envelope(ax, x, data, color, label, alpha=0.18):
    mean = data.mean(axis=0)
    lo   = data.min(axis=0)
    hi   = data.max(axis=0)
    ax.plot(x, mean, color=color, linewidth=2, label=label)
    ax.fill_between(x, lo, hi, color=color, alpha=alpha)

SAVEFIG_KW = dict(dpi=150, bbox_inches="tight")


# ══════════════════════════════════════════════════════════════
# 1. Collect REINFORCE data (Pendulum-v1)
# ══════════════════════════════════════════════════════════════

print("Running REINFORCE ...")
rf_train_r, rf_eval_r, rf_steps, rf_t_times, rf_w_times = [], [], [], [], []

for seed in SEEDS:
    env   = gym.make("Pendulum-v1")
    agent = REINFORCE(
        env                = env,
        seed               = seed,
        gamma              = 0.99,
        optimizerLR        = 1e-3,
        MAX_TRAIN_EPISODES = 150,
        MAX_EVAL_EPISODES  = 10,
    )
    train_r, eval_r, steps, t_times, w_times = agent.runREINFORCE()
    rf_train_r.append(train_r)
    rf_eval_r.append([r for (_, r) in eval_r])
    rf_steps.append(steps)
    rf_t_times.append(t_times)
    rf_w_times.append(w_times)
    env.close()
    print(f"  REINFORCE seed {seed} done.")

rf_train_r      = np.array(rf_train_r)
rf_eval_r       = np.array(rf_eval_r)
rf_steps_cs     = np.cumsum(np.array(rf_steps), axis=1)
rf_t_times      = np.array(rf_t_times)
rf_w_times      = np.array(rf_w_times)
rf_train_eps    = np.arange(1, rf_train_r.shape[1] + 1)
rf_eval_eps     = np.arange(EVAL_INTERVAL,
                             rf_train_r.shape[1] + 1, EVAL_INTERVAL)


# ══════════════════════════════════════════════════════════════
# 2. Collect VPG data (Pendulum-v1)
# ══════════════════════════════════════════════════════════════

print("Running VPG ...")
vpg_train_r, vpg_eval_r, vpg_steps, vpg_t_times, vpg_w_times = [], [], [], [], []

for seed in SEEDS:
    env   = gym.make("Pendulum-v1")
    agent = VPG(
        env                = env,
        seed               = seed,
        gamma              = 0.99,
        beta               = 0.01,
        optimizerFn        = optim.Adam,
        optimizerLR        = 1e-3,
        MAX_TRAIN_EPISODES = 150,
        MAX_EVAL_EPISODES  = 10,
    )
    train_r, eval_r, steps, t_times, w_times = agent.runVPG()
    vpg_train_r.append(train_r)
    vpg_eval_r.append([r for (_, r) in eval_r])
    vpg_steps.append(steps)
    vpg_t_times.append(t_times)
    vpg_w_times.append(w_times)
    env.close()
    print(f"  VPG seed {seed} done.")

vpg_train_r   = np.array(vpg_train_r)
vpg_eval_r    = np.array(vpg_eval_r)
vpg_steps_cs  = np.cumsum(np.array(vpg_steps), axis=1)
vpg_t_times   = np.array(vpg_t_times)
vpg_w_times   = np.array(vpg_w_times)
vpg_train_eps = np.arange(1, vpg_train_r.shape[1] + 1)
vpg_eval_eps  = np.arange(EVAL_INTERVAL,
                            vpg_train_r.shape[1] + 1, EVAL_INTERVAL)


# ══════════════════════════════════════════════════════════════
# 3. Collect DDPG data (HalfCheetah-v4)
# ══════════════════════════════════════════════════════════════

print("Running DDPG ...")
ddpg_train_r, ddpg_eval_r, ddpg_steps, ddpg_t_times, ddpg_w_times = [], [], [], [], []

for seed in SEEDS:
    agent = DDPG(
        env_name           = "HalfCheetah-v4",
        seed               = seed,
        gamma              = 0.99,
        tau                = 0.005,
        bufferSize         = 100_000,
        batch_size         = 256,
        updateFrequency    = 1,
        policyOptimizerFn  = optim.Adam,
        valueOptimizerFn   = optim.Adam,
        policyOptimizerLR  = 1e-3,
        valueOptimizerLR   = 1e-3,
        MAX_TRAIN_EPISODES = 150,
        MAX_EVAL_EPISODE   = 10,
    )
    train_r, eval_r, steps, t_times, w_times = agent.runDDPG()
    ddpg_train_r.append(train_r)
    ddpg_eval_r.append([r for (_, r) in eval_r])
    ddpg_steps.append(steps)
    ddpg_t_times.append(t_times)
    ddpg_w_times.append(w_times)
    agent.env.close()
    print(f"  DDPG seed {seed} done.")

ddpg_train_r   = np.array(ddpg_train_r)
ddpg_eval_r    = np.array(ddpg_eval_r)
ddpg_steps_cs  = np.cumsum(np.array(ddpg_steps), axis=1)
ddpg_t_times   = np.array(ddpg_t_times)
ddpg_w_times   = np.array(ddpg_w_times)
ddpg_train_eps = np.arange(1, ddpg_train_r.shape[1] + 1)
ddpg_eval_eps  = np.arange(EVAL_INTERVAL,
                             ddpg_train_r.shape[1] + 1, EVAL_INTERVAL)


# ══════════════════════════════════════════════════════════════
# 4. Collect TD3 data (HalfCheetah-v4)
# ══════════════════════════════════════════════════════════════

print("Running TD3 ...")
td3_train_r, td3_eval_r, td3_steps, td3_t_times, td3_w_times = [], [], [], [], []

for seed in SEEDS:
    agent = TD3(
        env_name             = "HalfCheetah-v4",
        seed                 = seed,
        gamma                = 0.99,
        tau                  = 0.005,
        bufferSize           = 100_000,
        updateFrequencyPolicy= 2,
        updateFrequencyValue = 2,
        trainPolicyFrequency = 2,
        policyOptimizerFn    = optim.Adam,
        valueOptimizerFn     = optim.Adam,
        policyOptimizerLR    = 3e-4,
        valueOptimizerLR     = 3e-4,
        MAX_TRAIN_EPISODES   = 150,
        MAX_EVAL_EPISODE     = 10,
        batch_size           = 256,
    )
    train_r, eval_r, steps, t_times, w_times = agent.runTD3()
    td3_train_r.append(train_r)
    td3_eval_r.append([r for (_, r) in eval_r])
    td3_steps.append(steps)
    td3_t_times.append(t_times)
    td3_w_times.append(w_times)
    agent.env.close()
    print(f"  TD3 seed {seed} done.")

td3_train_r   = np.array(td3_train_r)
td3_eval_r    = np.array(td3_eval_r)
td3_steps_cs  = np.cumsum(np.array(td3_steps), axis=1)
td3_t_times   = np.array(td3_t_times)
td3_w_times   = np.array(td3_w_times)
td3_train_eps = np.arange(1, td3_train_r.shape[1] + 1)
td3_eval_eps  = np.arange(EVAL_INTERVAL,
                            td3_train_r.shape[1] + 1, EVAL_INTERVAL)


# ══════════════════════════════════════════════════════════════
# 5. Collect PPO data (HalfCheetah-v4)
# ══════════════════════════════════════════════════════════════

print("Running PPO ...")
ppo_train_r, ppo_steps, ppo_t_times, ppo_w_times = [], [], [], []

for seed in SEEDS:
    train_rewards, steps_log, t_times, w_times, _ = train_ppo(
        env_id          = "HalfCheetah-v4",
        seed            = seed,
        total_timesteps = 150_000,
    )
    ppo_train_r.append(train_rewards)
    ppo_steps.append(steps_log)      # already cumulative
    ppo_t_times.append(t_times)
    ppo_w_times.append(w_times)
    print(f"  PPO seed {seed} done.")

min_ppo     = min(len(r) for r in ppo_train_r)
ppo_train_r = np.array([r[:min_ppo] for r in ppo_train_r])
ppo_steps   = np.array([s[:min_ppo] for s in ppo_steps])
ppo_t_times = np.array([t[:min_ppo] for t in ppo_t_times])
ppo_w_times = np.array([w[:min_ppo] for w in ppo_w_times])
ppo_update_idx = np.arange(1, min_ppo + 1)


# ══════════════════════════════════════════════════════════════
# Helper: normalise x-axis length for overlay
# All episode-based agents may have different episode counts.
# We normalise to a 0-1 progress axis for fair comparison.
# ══════════════════════════════════════════════════════════════

def to_progress(x):
    """Normalise x to [0, 1] range."""
    return (x - x[0]) / (x[-1] - x[0]) if x[-1] != x[0] else x


# ══════════════════════════════════════════════════════════════
# Plot 1 — Mean train reward vs episode
# ══════════════════════════════════════════════════════════════

fig, ax = plt.subplots(figsize=(10, 5))

plot_envelope(ax, rf_train_eps,
              rolling_mean(rf_train_r, 20),
              COLORS["REINFORCE"], "REINFORCE")
plot_envelope(ax, vpg_train_eps,
              rolling_mean(vpg_train_r, 20),
              COLORS["VPG"], "VPG")
plot_envelope(ax, ddpg_train_eps,
              rolling_mean(ddpg_train_r, 20),
              COLORS["DDPG"], "DDPG")
plot_envelope(ax, td3_train_eps,
              rolling_mean(td3_train_r, 20),
              COLORS["TD3"], "TD3")
plot_envelope(ax, ppo_update_idx,
              rolling_mean(ppo_train_r, 20),
              COLORS["PPO"], "PPO")

ax.set_title("Mean train reward vs episode — all agents")
ax.set_xlabel("Episode / Update")
ax.set_ylabel("Train reward (rolling mean, w=20)")
ax.legend(fontsize=9)
ax.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("compare_plot1_train_reward.png", **SAVEFIG_KW)
plt.show()
plt.close()


# ══════════════════════════════════════════════════════════════
# Plot 2 — Mean eval reward vs episode
# ══════════════════════════════════════════════════════════════

fig, ax = plt.subplots(figsize=(10, 5))

plot_envelope(ax, rf_eval_eps,   rf_eval_r,   COLORS["REINFORCE"], "REINFORCE")
plot_envelope(ax, vpg_eval_eps,  vpg_eval_r,  COLORS["VPG"],       "VPG")
plot_envelope(ax, ddpg_eval_eps, ddpg_eval_r, COLORS["DDPG"],      "DDPG")
plot_envelope(ax, td3_eval_eps,  td3_eval_r,  COLORS["TD3"],       "TD3")
# PPO uses train reward as proxy (no separate eval loop)
plot_envelope(ax, ppo_update_idx, ppo_train_r, COLORS["PPO"],      "PPO (proxy)")

ax.set_title("Mean eval reward vs episode — all agents")
ax.set_xlabel("Episode / Update")
ax.set_ylabel("Eval reward")
ax.legend(fontsize=9)
ax.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("compare_plot2_eval_reward.png", **SAVEFIG_KW)
plt.show()
plt.close()


# ══════════════════════════════════════════════════════════════
# Plot 3 — Total steps vs episode
# ══════════════════════════════════════════════════════════════

fig, ax = plt.subplots(figsize=(10, 5))

plot_envelope(ax, rf_train_eps,    rf_steps_cs,   COLORS["REINFORCE"], "REINFORCE")
plot_envelope(ax, vpg_train_eps,   vpg_steps_cs,  COLORS["VPG"],       "VPG")
plot_envelope(ax, ddpg_train_eps,  ddpg_steps_cs, COLORS["DDPG"],      "DDPG")
plot_envelope(ax, td3_train_eps,   td3_steps_cs,  COLORS["TD3"],       "TD3")
plot_envelope(ax, ppo_update_idx,  ppo_steps,     COLORS["PPO"],       "PPO")

ax.set_title("Total steps vs episode — all agents")
ax.set_xlabel("Episode / Update")
ax.set_ylabel("Cumulative steps")
ax.legend(fontsize=9)
ax.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("compare_plot3_total_steps.png", **SAVEFIG_KW)
plt.show()
plt.close()


# ══════════════════════════════════════════════════════════════
# Plot 4 — Train time vs episode
# ══════════════════════════════════════════════════════════════

fig, ax = plt.subplots(figsize=(10, 5))

plot_envelope(ax, rf_train_eps,   rf_t_times,   COLORS["REINFORCE"], "REINFORCE")
plot_envelope(ax, vpg_train_eps,  vpg_t_times,  COLORS["VPG"],       "VPG")
plot_envelope(ax, ddpg_train_eps, ddpg_t_times, COLORS["DDPG"],      "DDPG")
plot_envelope(ax, td3_train_eps,  td3_t_times,  COLORS["TD3"],       "TD3")
plot_envelope(ax, ppo_update_idx, ppo_t_times,  COLORS["PPO"],       "PPO")

ax.set_title("Train time vs episode — all agents")
ax.set_xlabel("Episode / Update")
ax.set_ylabel("Cumulative train time (s)")
ax.legend(fontsize=9)
ax.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("compare_plot4_train_time.png", **SAVEFIG_KW)
plt.show()
plt.close()


# ══════════════════════════════════════════════════════════════
# Plot 5 — Wall clock time vs episode
# ══════════════════════════════════════════════════════════════

fig, ax = plt.subplots(figsize=(10, 5))

plot_envelope(ax, rf_train_eps,   rf_w_times,   COLORS["REINFORCE"], "REINFORCE")
plot_envelope(ax, vpg_train_eps,  vpg_w_times,  COLORS["VPG"],       "VPG")
plot_envelope(ax, ddpg_train_eps, ddpg_w_times, COLORS["DDPG"],      "DDPG")
plot_envelope(ax, td3_train_eps,  td3_w_times,  COLORS["TD3"],       "TD3")
plot_envelope(ax, ppo_update_idx, ppo_w_times,  COLORS["PPO"],       "PPO")

ax.set_title("Wall clock time vs episode — all agents")
ax.set_xlabel("Episode / Update")
ax.set_ylabel("Wall clock time (s)")
ax.legend(fontsize=9)
ax.grid(True, linewidth=0.4, alpha=0.5)
plt.tight_layout()
plt.savefig("compare_plot5_wall_time.png", **SAVEFIG_KW)
plt.show()
plt.close()
````
